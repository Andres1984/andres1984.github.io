```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

n = 500

# -----------------------------------
# 1. ASSOCIATION (confounding)
# -----------------------------------
Z = np.random.normal(0, 1, n)  # hidden variable
X_assoc = 2*Z + np.random.normal(0, 1, n)
Y_assoc = 3*Z + np.random.normal(0, 1, n)

plt.figure()
plt.scatter(X_assoc, Y_assoc)
plt.title("Association (Driven by a Hidden Variable)")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()


# -----------------------------------
# 2. CORRELATION (linear relationship)
# -----------------------------------
X_corr = np.random.normal(0, 1, n)
Y_corr = 2 * X_corr + np.random.normal(0, 0.5, n)

plt.figure()
plt.scatter(X_corr, Y_corr)
plt.title("Correlation (Linear Relationship)")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()


# -----------------------------------
# 3. CAUSALITY (data generating process)
# -----------------------------------
X_causal = np.random.normal(0, 1, n)
Y_causal = 2 * X_causal + np.random.normal(0, 1, n)

# Intervention: shift X
X_intervene = X_causal + 2
Y_intervene = 2 * X_intervene + np.random.normal(0, 1, n)

plt.figure()
plt.scatter(X_causal, Y_causal)
plt.scatter(X_intervene, Y_intervene)
plt.title("Causality (Effect of Intervention on X)")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

# =========================
# 1) ASSOCIATION: Ice cream and drownings
# Z = temperature
# =========================
n = 300
temperature = np.random.normal(25, 5, n)

ice_cream = 20 + 3.5 * temperature + np.random.normal(0, 4, n)
drownings = 2 + 0.6 * temperature + np.random.normal(0, 1.5, n)

# Residualize against temperature
coef_x = np.polyfit(temperature, ice_cream, 1)
x_hat = np.polyval(coef_x, temperature)
ice_resid = ice_cream - x_hat

coef_y = np.polyfit(temperature, drownings, 1)
y_hat = np.polyval(coef_y, temperature)
drown_resid = drownings - y_hat

# =========================
# 2) CORRELATION: Celsius and Fahrenheit
# =========================
celsius = np.linspace(-10, 40, 200)
fahrenheit = 32 + 9/5 * celsius

# =========================
# 3) CAUSALITY: Fertilizer and plant growth
# randomized treatment
# =========================
n2 = 200
treatment = np.random.binomial(1, 0.5, n2)   # randomized assignment
growth = 10 + 4 * treatment + np.random.normal(0, 2, n2)

# =========================
# FIGURE 1: ASSOCIATION
# =========================
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Raw association
axes[0].scatter(ice_cream, drownings, alpha=0.7)
axes[0].set_title("Raw association")
axes[0].set_xlabel("Ice cream sales")
axes[0].set_ylabel("Drowning incidents")

# Colored by temperature
sc = axes[1].scatter(ice_cream, drownings, c=temperature, alpha=0.7)
axes[1].set_title("Association changes once Z is visible")
axes[1].set_xlabel("Ice cream sales")
axes[1].set_ylabel("Drowning incidents")
cbar = plt.colorbar(sc, ax=axes[1])
cbar.set_label("Temperature")

# Residualized
axes[2].scatter(ice_resid, drown_resid, alpha=0.7)
axes[2].set_title("After adjusting for temperature")
axes[2].set_xlabel("Ice cream residuals")
axes[2].set_ylabel("Drowning residuals")

plt.tight_layout()
plt.show()

# =========================
# FIGURE 2: CORRELATION
# =========================
plt.figure(figsize=(6, 4))
plt.plot(celsius, fahrenheit)
plt.title("Correlation: Celsius vs Fahrenheit")
plt.xlabel("Celsius")
plt.ylabel("Fahrenheit")
plt.tight_layout()
plt.show()

# =========================
# FIGURE 3: CAUSALITY
# =========================
plt.figure(figsize=(6, 4))
plt.boxplot(
    [growth[treatment == 0], growth[treatment == 1]],
    tick_labels=["Control", "Fertilizer"]
)
plt.title("Causality: randomized intervention")
plt.ylabel("Plant growth")
plt.tight_layout()
plt.show()

# Optional printed correlations
print("Raw association (ice cream, drownings):",
      np.corrcoef(ice_cream, drownings)[0, 1])

print("Adjusted association (residuals):",
      np.corrcoef(ice_resid, drown_resid)[0, 1])

print("Correlation (Celsius, Fahrenheit):",
      np.corrcoef(celsius, fahrenheit)[0, 1])

print("Mean growth - control:", growth[treatment == 0].mean())
print("Mean growth - fertilizer:", growth[treatment == 1].mean())
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

# -----------------------------
# Simulated example
# X: exercise hours
# Y: cholesterol
# Z: age
# -----------------------------
n = 400

age = np.random.uniform(20, 80, n)

# Older people tend to exercise a bit more (doctor advice, lifestyle change, etc.)
exercise = 1 + 0.08 * age + np.random.normal(0, 1.2, n)
exercise = np.clip(exercise, 0, None)

# True within-age effect: more exercise lowers cholesterol
cholesterol = 140 + 1.4 * age - 6.0 * exercise + np.random.normal(0, 8, n)

df = pd.DataFrame({
    "age": age,
    "exercise": exercise,
    "cholesterol": cholesterol
})

# Age groups
df["age_group"] = pd.cut(
    df["age"],
    bins=[20, 35, 50, 65, 80],
    labels=["20-35", "36-50", "51-65", "66-80"],
    include_lowest=True
)

# -----------------------------
# Graph 1: naive relationship
# -----------------------------
plt.figure(figsize=(6,4))
plt.scatter(df["exercise"], df["cholesterol"], alpha=0.7)

# overall regression line
b1, b0 = np.polyfit(df["exercise"], df["cholesterol"], 1)
xline = np.linspace(df["exercise"].min(), df["exercise"].max(), 100)
plt.plot(xline, b1 * xline + b0)

plt.xlabel("Hours of exercise per week")
plt.ylabel("Cholesterol")
plt.title("Observed relationship: exercise and cholesterol")
plt.tight_layout()
plt.show()

# -----------------------------
# Graph 2: relationship by age group
# -----------------------------
plt.figure(figsize=(7,5))

for group in df["age_group"].cat.categories:
    dfg = df[df["age_group"] == group]
    plt.scatter(dfg["exercise"], dfg["cholesterol"], alpha=0.7, label=f"Age {group}")
    
    # regression line within each age group
    m, c = np.polyfit(dfg["exercise"], dfg["cholesterol"], 1)
    xg = np.linspace(dfg["exercise"].min(), dfg["exercise"].max(), 100)
    plt.plot(xg, m * xg + c)

plt.xlabel("Hours of exercise per week")
plt.ylabel("Cholesterol")
plt.title("After accounting for age")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

n = 400

# -----------------------------
# TRUE SIGNAL (causal structure)
# -----------------------------
X_true = np.random.normal(0, 1, n)
Y_true = 2 * X_true + np.random.normal(0, 0.5, n)

# -----------------------------
# NOISE VARIABLES (irrelevant)
# -----------------------------
noise1 = np.random.normal(0, 3, n)
noise2 = np.random.normal(0, 3, n)

# Observed X (contaminated with noise)
X_noisy = X_true + noise1

# -----------------------------
# PLOT 1: ML without causal inference
# -----------------------------
plt.figure(figsize=(6,4))
plt.scatter(X_noisy, Y_true, alpha=0.4, label="Noise + signal")

# highlight points used (random selection mimicking model confusion)
idx = np.random.choice(range(n), size=80, replace=False)
plt.scatter(X_noisy[idx], Y_true[idx], label="Selected by model")

plt.title("Machine Learning without causal structure")
plt.xlabel("Observed X (noisy)")
plt.ylabel("Y")
plt.legend()
plt.tight_layout()
plt.savefig("ml_no_causal.png")
plt.show()


# -----------------------------
# PLOT 2: WITH CAUSAL INFERENCE
# -----------------------------
plt.figure(figsize=(6,4))
plt.scatter(X_true, Y_true, alpha=0.4, label="True signal")

# highlight points aligned with causal structure
mask = np.abs(Y_true - 2*X_true) < 1
plt.scatter(X_true[mask], Y_true[mask], label="Causal signal")

plt.title("Machine Learning with causal structure")
plt.xlabel("Causal X")
plt.ylabel("Y")
plt.legend()
plt.tight_layout()
plt.savefig("ml_causal.png")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# -----------------------------
# Generate data
# -----------------------------
n_main = 260
x_main = np.random.uniform(0, 10, n_main)
y_main = 2 + 1.2 * x_main + np.random.normal(0, 1.2, n_main)

# Spurious noisy cluster: looks predictive locally but is not structural
n_spur = 90
x_spur = np.random.normal(7.5, 0.6, n_spur)
y_spur = np.random.normal(14.5, 0.8, n_spur)

# Random background noise
n_noise = 80
x_noise = np.random.uniform(0, 10, n_noise)
y_noise = np.random.uniform(0, 18, n_noise)

# Full observed dataset
x = np.concatenate([x_main, x_spur, x_noise])
y = np.concatenate([y_main, y_spur, y_noise])

# -----------------------------
# Figure 1: ML without causal inference
# Highlights too much structure, including spurious local cluster
# -----------------------------
plt.figure(figsize=(6, 4))

# all data
plt.scatter(x, y, alpha=0.25, s=22)

# highlight "selected" points:
# main trend + spurious cluster, mimicking a model chasing all predictive patterns
mask_ml = (
    (np.abs(y - (2 + 1.2 * x)) < 1.0) |   # true stable trend
    (((x - 7.5)**2 + (y - 14.5)**2) < 1.8**2)  # local spurious cluster
)

plt.scatter(x[mask_ml], y[mask_ml], s=24)

# draw both lines/patterns
xline = np.linspace(0, 10, 200)
plt.plot(xline, 2 + 1.2 * xline, linewidth=2)

# local spurious fit
xspur_line = np.linspace(6.2, 8.8, 100)
plt.plot(xspur_line, np.full_like(xspur_line, 14.5), linewidth=2, linestyle="--")

plt.title("Machine Learning without Causal Inference")
plt.xlabel("X")
plt.ylabel("Y")
plt.tight_layout()
plt.savefig("ml_without_causal_structure.png", dpi=200)
plt.show()

# -----------------------------
# Figure 2: ML with causal inference
# Focuses on the stable pattern only
# -----------------------------
plt.figure(figsize=(6, 4))

plt.scatter(x, y, alpha=0.25, s=22)

# highlight only points near the stable global pattern
mask_causal = np.abs(y - (2 + 1.2 * x)) < 1.0
plt.scatter(x[mask_causal], y[mask_causal], s=24)

plt.plot(xline, 2 + 1.2 * xline, linewidth=2)

plt.title("Machine Learning with Causal Structure")
plt.xlabel("X")
plt.ylabel("Y")
plt.tight_layout()
plt.savefig("ml_with_causal_structure.png", dpi=200)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

n = 250

# -------------------------
# Story 1: X -> Y
# -------------------------
X1 = np.random.normal(0, 1, n)
Y1 = 1.8 * X1 + np.random.normal(0, 0.9, n)

# -------------------------
# Story 2: Y -> X
# -------------------------
Y2 = np.random.normal(0, 1, n)
X2 = 1.8 * Y2 + np.random.normal(0, 0.9, n)

# -------------------------
# Figure
# -------------------------
fig, axes = plt.subplots(1, 3, figsize=(14, 4.2))

# Panel 1: observed association
axes[0].scatter(X1, Y1, alpha=0.65, s=28, color="#444444")
coef = np.polyfit(X1, Y1, 1)
xline = np.linspace(X1.min(), X1.max(), 100)
axes[0].plot(xline, np.poly1d(coef)(xline), linewidth=2.5, color="#006837")
axes[0].set_title("Observed data", fontsize=13, weight="bold")
axes[0].set_xlabel("X")
axes[0].set_ylabel("Y")
axes[0].text(
    0.05, 0.92,
    "We observe association",
    transform=axes[0].transAxes,
    fontsize=10,
    bbox=dict(boxstyle="round,pad=0.25", fc="white", ec="0.7")
)

# Panel 2: X causes Y
axes[1].scatter(X1, Y1, alpha=0.65, s=28, color="#444444")
coef1 = np.polyfit(X1, Y1, 1)
axes[1].plot(xline, np.poly1d(coef1)(xline), linewidth=2.5, color="#006837")
axes[1].set_title("Possible story 1: X → Y", fontsize=13, weight="bold")
axes[1].set_xlabel("X")
axes[1].set_ylabel("Y")
axes[1].text(
    0.05, 0.86,
    "Step 1: generate X\nStep 2: Y responds to X",
    transform=axes[1].transAxes,
    fontsize=10,
    bbox=dict(boxstyle="round,pad=0.3", fc="#eef7f1", ec="#006837")
)

# Panel 3: Y causes X
axes[2].scatter(X2, Y2, alpha=0.65, s=28, color="#444444")
coef2 = np.polyfit(X2, Y2, 1)
xline2 = np.linspace(X2.min(), X2.max(), 100)
axes[2].plot(xline2, np.poly1d(coef2)(xline2), linewidth=2.5, color="#B8860B")
axes[2].set_title("Possible story 2: Y → X", fontsize=13, weight="bold")
axes[2].set_xlabel("X")
axes[2].set_ylabel("Y")
axes[2].text(
    0.05, 0.86,
    "Step 1: generate Y\nStep 2: X responds to Y",
    transform=axes[2].transAxes,
    fontsize=10,
    bbox=dict(boxstyle="round,pad=0.3", fc="#fbf6e8", ec="#B8860B")
)

# visual cleanup
for ax in axes:
    ax.grid(alpha=0.15)
    ax.set_xticks([])
    ax.set_yticks([])
    for spine in ax.spines.values():
        spine.set_alpha(0.3)

plt.suptitle("Direction is a property of the causal story, not of the scatter plot", fontsize=15, weight="bold")
plt.tight_layout()
plt.savefig("directed_story.png", dpi=300, transparent=True, bbox_inches="tight")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Reproducibilidad
np.random.seed(42)

# Variable oculta: temperatura
n = 300
temperature = np.random.normal(loc=30, scale=5, size=n)

# Variables observadas afectadas por temperatura
ice_cream_sales = 20 + 3.5 * temperature + np.random.normal(0, 8, n)
drownings = -15 + 1.8 * temperature + np.random.normal(0, 5, n)

# Figura 1: correlación espuria + verdadera causa
fig = plt.figure(figsize=(12, 5))

# -----------------------------
# Panel izquierdo: correlación aparente
# -----------------------------
ax1 = fig.add_subplot(1, 2, 1)
ax1.scatter(ice_cream_sales, drownings, alpha=0.75)
ax1.set_title("Observed data")
ax1.set_xlabel("Ice cream sales")
ax1.set_ylabel("Drownings")

# línea de tendencia
coef = np.polyfit(ice_cream_sales, drownings, 1)
x_line = np.linspace(ice_cream_sales.min(), ice_cream_sales.max(), 100)
y_line = coef[0] * x_line + coef[1]
ax1.plot(x_line, y_line)

ax1.text(
    0.05, 0.92,
    "Looks causal:\nmore ice cream → more drownings",
    transform=ax1.transAxes,
    fontsize=10,
    verticalalignment='top',
    bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="gray")
)

# -----------------------------
# Panel derecho: estructura causal real
# -----------------------------
ax2 = fig.add_subplot(1, 2, 2)
ax2.axis("off")
ax2.set_title("True causal structure")

# posiciones de nodos
pos = {
    "Temperature": (0.5, 0.8),
    "Ice cream sales": (0.2, 0.3),
    "Drownings": (0.8, 0.3)
}

# dibujar nodos
for label, (x, y) in pos.items():
    circle = plt.Circle((x, y), 0.12, fill=False, linewidth=2)
    ax2.add_patch(circle)
    ax2.text(x, y, label, ha='center', va='center', fontsize=10)

# dibujar flechas
ax2.annotate(
    "", xy=pos["Ice cream sales"], xytext=pos["Temperature"],
    arrowprops=dict(arrowstyle="->", lw=2)
)
ax2.annotate(
    "", xy=pos["Drownings"], xytext=pos["Temperature"],
    arrowprops=dict(arrowstyle="->", lw=2)
)

ax2.text(
    0.5, 0.05,
    "Temperature causes both variables.\nThe observed correlation is not causal.",
    ha="center",
    fontsize=10
)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------
# Collider example: Smart Cities
# X = Traffic
# Y = Incidents
# Z = Emergency response activation
# ---------------------------------

np.random.seed(42)
n = 2000

# In the full population, X and Y are independent
traffic = np.random.normal(0, 1, n)
incidents = np.random.normal(0, 1, n)

# Collider: response depends on BOTH traffic and incidents
# Higher traffic and/or higher incidents increase the chance of response
latent_response = 0.9 * traffic + 0.9 * incidents + np.random.normal(0, 0.8, n)

# Condition on the collider:
# keep only observations where emergency response was activated
threshold = np.quantile(latent_response, 0.7)
selected = latent_response > threshold

traffic_sel = traffic[selected]
incidents_sel = incidents[selected]

# Correlations for annotation
corr_full = np.corrcoef(traffic, incidents)[0, 1]
corr_sel = np.corrcoef(traffic_sel, incidents_sel)[0, 1]

# ---------------------------------
# Plot
# ---------------------------------
fig = plt.figure(figsize=(12, 5))

# BEFORE conditioning
ax1 = fig.add_subplot(1, 2, 1)
ax1.scatter(traffic, incidents, alpha=0.35)
ax1.set_title("Before conditioning")
ax1.set_xlabel("Traffic")
ax1.set_ylabel("Incidents")
ax1.text(
    0.04, 0.95,
    f"Corr ≈ {corr_full:.2f}\nNo clear relationship",
    transform=ax1.transAxes,
    va="top",
    bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="gray")
)

# AFTER conditioning on collider
ax2 = fig.add_subplot(1, 2, 2)
ax2.scatter(traffic_sel, incidents_sel, alpha=0.5)
ax2.set_title("After conditioning on emergency response")
ax2.set_xlabel("Traffic")
ax2.set_ylabel("Incidents")

# Trend line
coef = np.polyfit(traffic_sel, incidents_sel, 1)
x_line = np.linspace(traffic_sel.min(), traffic_sel.max(), 100)
y_line = coef[0] * x_line + coef[1]
ax2.plot(x_line, y_line, linewidth=2)

ax2.text(
    0.04, 0.95,
    f"Corr ≈ {corr_sel:.2f}\nFalse dependence appears",
    transform=ax2.transAxes,
    va="top",
    bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="gray")
)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# COLLIDER EXAMPLE: Smart Cities
# X = Traffic
# Y = Incidents
# Z = Emergency response activation
#
# Goal:
# 1. Show X and Y are unrelated in the full population
# 2. Show that conditioning on Z creates a false relationship
# 3. Make Z visible through:
#    - color intensity
#    - a selection boundary
#    - a shaded selected region
# ------------------------------------------------------------

np.random.seed(42)
n = 2500

# -----------------------------
# 1. Generate independent causes
# -----------------------------
traffic = np.random.normal(0, 1, n)      # X
incidents = np.random.normal(0, 1, n)    # Y

# ---------------------------------------------------
# 2. Collider:
#    Emergency response depends on BOTH X and Y
# ---------------------------------------------------
# Think of this as a latent score that triggers response
latent_response = 0.9 * traffic + 0.9 * incidents + np.random.normal(0, 0.7, n)

# Threshold for "response activated"
threshold = np.quantile(latent_response, 0.72)
selected = latent_response > threshold

traffic_sel = traffic[selected]
incidents_sel = incidents[selected]
response_sel = latent_response[selected]

# Correlations for annotation
corr_full = np.corrcoef(traffic, incidents)[0, 1]
corr_sel = np.corrcoef(traffic_sel, incidents_sel)[0, 1]

# ------------------------------------------
# 3. Selection boundary for visualization
# ------------------------------------------
# Approximate boundary using the deterministic part:
# 0.9*X + 0.9*Y = threshold
# => Y = (threshold - 0.9*X) / 0.9
x_line = np.linspace(traffic.min() - 0.5, traffic.max() + 0.5, 300)
y_boundary = (threshold - 0.9 * x_line) / 0.9

# ------------------------------------------
# 4. Figure
# ------------------------------------------
fig = plt.figure(figsize=(14, 6))

# ==========================================================
# LEFT PANEL: Full population + visible effect of Z
# ==========================================================
ax1 = fig.add_subplot(1, 2, 1)

# Scatter with color showing the latent collider score Z
sc1 = ax1.scatter(
    traffic,
    incidents,
    c=latent_response,
    alpha=0.45
)

# Selection boundary
ax1.plot(
    x_line,
    y_boundary,
    linestyle="--",
    linewidth=2
)

# Shade the selected region (roughly where Z is large)
y_top = np.full_like(x_line, incidents.max() + 0.8)
ax1.fill_between(
    x_line,
    y_boundary,
    y_top,
    alpha=0.12
)

ax1.set_title("Before conditioning")
ax1.set_xlabel("Traffic")
ax1.set_ylabel("Incidents")

ax1.text(
    0.03, 0.96,
    f"Corr ≈ {corr_full:.2f}\nTraffic and incidents are independent",
    transform=ax1.transAxes,
    va="top",
    bbox=dict(boxstyle="round,pad=0.35", fc="white", ec="gray")
)

ax1.text(
    0.03, 0.14,
    "Dashed line: response threshold\nShaded area: observations with high response activation",
    transform=ax1.transAxes,
    va="bottom",
    bbox=dict(boxstyle="round,pad=0.35", fc="white", ec="gray")
)

# Optional colorbar to show Z
cbar1 = plt.colorbar(sc1, ax=ax1)
cbar1.set_label("Latent response score (Z)")

# ==========================================================
# RIGHT PANEL: After conditioning on collider Z
# ==========================================================
ax2 = fig.add_subplot(1, 2, 2)

sc2 = ax2.scatter(
    traffic_sel,
    incidents_sel,
    c=response_sel,
    alpha=0.65
)

# Trend line after conditioning
coef = np.polyfit(traffic_sel, incidents_sel, 1)
x_fit = np.linspace(traffic_sel.min(), traffic_sel.max(), 200)
y_fit = coef[0] * x_fit + coef[1]
ax2.plot(
    x_fit,
    y_fit,
    linewidth=2
)

ax2.set_title("After conditioning on emergency response")
ax2.set_xlabel("Traffic")
ax2.set_ylabel("Incidents")

ax2.text(
    0.03, 0.96,
    f"Corr ≈ {corr_sel:.2f}\nA false dependence appears",
    transform=ax2.transAxes,
    va="top",
    bbox=dict(boxstyle="round,pad=0.35", fc="white", ec="gray")
)

ax2.text(
    0.03, 0.14,
    "Only observations above the response threshold remain",
    transform=ax2.transAxes,
    va="bottom",
    bbox=dict(boxstyle="round,pad=0.35", fc="white", ec="gray")
)

cbar2 = plt.colorbar(sc2, ax=ax2)
cbar2.set_label("Latent response score (Z)")

plt.tight_layout()

# Save for LaTeX
plt.savefig("collider_smartcity.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# MEDIATOR EXAMPLE: Smart Cities
# X = Rain
# Z = Road slipperiness (mediator)
# Y = Accidents
#
# Goal:
# 1. Show the total relationship between rain and accidents
# 2. Show that after accounting for the mediator, the relationship weakens
# 3. Make Z visible through color
# ------------------------------------------------------------

np.random.seed(42)
n = 700

# -----------------------------
# 1. Explanatory variable: Rain
# -----------------------------
rain = np.random.normal(loc=0.8, scale=0.6, size=n)
rain = np.clip(rain, -0.5, 2.2)

# ------------------------------------------
# 2. Mediator: Road slipperiness
#    Rain affects road condition
# ------------------------------------------
slipperiness = 1.2 * rain + np.random.normal(0, 0.35, n)

# ---------------------------------------------------
# 3. Outcome: Accidents
#    Accidents depend strongly on slipperiness
#    and weakly/directly on rain
# ---------------------------------------------------
accidents = 18 + 7.5 * slipperiness + 1.8 * rain + np.random.normal(0, 3.2, n)

# ---------------------------------------------------
# 4. "Adjustment" for the mediator
#    We remove the linear effect of slipperiness from accidents
#    This leaves a residualized outcome:
#    accidents_after ~ direct effect of rain
# ---------------------------------------------------
coef_z = np.polyfit(slipperiness, accidents, 1)
accidents_adjusted = accidents - (coef_z[0] * slipperiness)

# Shift for cleaner visualization
accidents_adjusted = accidents_adjusted - accidents_adjusted.min() + 5

# Correlations
corr_before = np.corrcoef(rain, accidents)[0, 1]
corr_after = np.corrcoef(rain, accidents_adjusted)[0, 1]

# ---------------------------------------------------
# 5. Categorize mediator for legend-like interpretation
# ---------------------------------------------------
q1, q2, q3 = np.quantile(slipperiness, [0.25, 0.50, 0.75])

labels = np.empty(n, dtype=object)
labels[slipperiness <= q1] = "Low slipperiness"
labels[(slipperiness > q1) & (slipperiness <= q2)] = "Moderate slipperiness"
labels[(slipperiness > q2) & (slipperiness <= q3)] = "High slipperiness"
labels[slipperiness > q3] = "Very high slipperiness"

color_map = {
    "Low slipperiness": "#2E86C1",
    "Moderate slipperiness": "#76B7C4",
    "High slipperiness": "#F0B44D",
    "Very high slipperiness": "#E76F51",
}

# ---------------------------------------------------
# 6. Create figure
# ---------------------------------------------------
fig = plt.figure(figsize=(13, 6))

# =========================
# LEFT PANEL: Before adjustment
# =========================
ax1 = fig.add_subplot(1, 2, 1)

for lab in color_map:
    idx = labels == lab
    ax1.scatter(
        rain[idx],
        accidents[idx],
        alpha=0.65,
        label=lab
    )

# Trend line
coef_before = np.polyfit(rain, accidents, 1)
x_line = np.linspace(rain.min(), rain.max(), 200)
y_line_before = coef_before[0] * x_line + coef_before[1]
ax1.plot(x_line, y_line_before, linewidth=2.5)

ax1.set_title("Before adjustment")
ax1.set_xlabel("Rain")
ax1.set_ylabel("Accidents")

ax1.text(
    0.05, 0.92,
    f"Corr ≈ {corr_before:.2f}\nTotal relationship",
    transform=ax1.transAxes,
    va="top",
    bbox=dict(boxstyle="round,pad=0.35", fc="white", ec="gray")
)

# =========================
# RIGHT PANEL: After adjustment
# =========================
ax2 = fig.add_subplot(1, 2, 2)

for lab in color_map:
    idx = labels == lab
    ax2.scatter(
        rain[idx],
        accidents_adjusted[idx],
        alpha=0.65,
        label=lab
    )

# Trend line
coef_after = np.polyfit(rain, accidents_adjusted, 1)
y_line_after = coef_after[0] * x_line + coef_after[1]
ax2.plot(x_line, y_line_after, linewidth=2.5)

ax2.set_title("After adjusting for road condition")
ax2.set_xlabel("Rain")
ax2.set_ylabel("Accidents")

ax2.text(
    0.05, 0.92,
    f"Corr ≈ {corr_after:.2f}\nDirect relationship",
    transform=ax2.transAxes,
    va="top",
    bbox=dict(boxstyle="round,pad=0.35", fc="white", ec="gray")
)

# Legend on right panel
handles, leg_labels = ax2.get_legend_handles_labels()
by_label = dict(zip(leg_labels, handles))
ax2.legend(
    by_label.values(),
    by_label.keys(),
    loc="lower right",
    frameon=True
)

plt.tight_layout()

# Save image for LaTeX
plt.savefig("mediator_smartcity.png", dpi=300, bbox_inches="tight")

plt.show()
```
