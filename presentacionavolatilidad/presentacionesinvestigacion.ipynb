```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(123)

fig, ax = plt.subplots(figsize=(10, 10), facecolor="black")
ax.set_facecolor("black")

# =========================================================
# FUNCIONES
# =========================================================
def ellipse_points(cx, cy, n, a, b, angle_deg=0):
    theta = np.random.uniform(0, 2*np.pi, n)
    r = np.sqrt(np.random.uniform(0, 1, n))
    x = a * r * np.cos(theta)
    y = b * r * np.sin(theta)

    ang = np.deg2rad(angle_deg)
    xr = x * np.cos(ang) - y * np.sin(ang)
    yr = x * np.sin(ang) + y * np.cos(ang)

    return cx + xr, cy + yr


def ring_points(cx, cy, n, a, b, angle_deg=0, rmin=0.86, rmax=1.0):
    theta = np.random.uniform(0, 2*np.pi, n)
    r = np.sqrt(np.random.uniform(rmin**2, rmax**2, n))

    x = a * r * np.cos(theta)
    y = b * r * np.sin(theta)

    ang = np.deg2rad(angle_deg)
    xr = x * np.cos(ang) - y * np.sin(ang)
    yr = x * np.sin(ang) + y * np.cos(ang)

    return cx + xr, cy + yr


def remove_inner_noise(x, y, centers, radius=1.55):
    mask = np.ones(len(x), dtype=bool)
    for cx, cy in centers:
        mask &= ((x - cx)**2 + (y - cy)**2) > radius**2
    return x[mask], y[mask]

# =========================================================
# CENTROS
# =========================================================
centers = [(-1.05, 1.05), (1.05, 1.05), (-1.05, -1.05), (1.05, -1.05)]
angles  = [15, -15, -10, 12]

# =========================================================
# RUIDO GLOBAL SOLO AFUERA
# =========================================================
n_bg = 12000
x_bg = np.random.normal(0, 2.6, n_bg)
y_bg = np.random.normal(0, 2.2, n_bg)

mask = (x_bg**2 / 4.0**2 + y_bg**2 / 3.6**2) < 1.0
x_bg, y_bg = x_bg[mask], y_bg[mask]
x_bg, y_bg = remove_inner_noise(x_bg, y_bg, centers, radius=1.55)

ax.scatter(
    x_bg, y_bg,
    s=5, c="#00cfff", alpha=0.45, edgecolors="none",
    label="Noise", zorder=1
)

# =========================================================
# MARKET PRICE MÁS PEQUEÑO
# =========================================================
x_mp, y_mp = ring_points(0, 0, 900, a=3.0, b=2.7, rmin=0.88, rmax=1.0)
ax.scatter(
    x_mp, y_mp,
    s=22, c="#ffb000", alpha=0.8, edgecolors="none",
    label="Market Price", zorder=3
)

# =========================================================
# SUBMERCADOS GRANDES, SIN RUIDO INTERNO
# =========================================================
for i, ((cx, cy), ang) in enumerate(zip(centers, angles)):

    # Long-Term Volatility (derecha)
    x_ltv, y_ltv = ellipse_points(cx + 0.28, cy, 380, 0.62, 0.50, ang)
    ax.scatter(
        x_ltv, y_ltv,
        s=34, c="yellow", alpha=0.97, edgecolors="none",
        label="Long-Term Volatility" if i == 0 else None,
        zorder=5
    )

    # Persistent Volatility (arriba izquierda)
    x_pv, y_pv = ellipse_points(cx - 0.34, cy + 0.30, 360, 0.58, 0.45, ang)
    ax.scatter(
        x_pv, y_pv,
        s=34, c="red", alpha=0.97, edgecolors="none",
        label="Persistent Volatility" if i == 0 else None,
        zorder=5
    )

    # Recurrent Shocks (abajo izquierda)
    x_rs, y_rs = ellipse_points(cx - 0.34, cy - 0.30, 360, 0.58, 0.45, ang)
    ax.scatter(
        x_rs, y_rs,
        s=34, c="green", alpha=0.97, edgecolors="none",
        label="Recurrent Shocks" if i == 0 else None,
        zorder=5
    )

    # Bias como borde local
    x_bias, y_bias = ring_points(cx, cy, 460, 1.35, 1.05, angle_deg=ang, rmin=0.86, rmax=1.0)
    ax.scatter(
        x_bias, y_bias,
        s=40, c="#d16cff", alpha=0.98, edgecolors="none",
        label="Bias" if i == 0 else None,
        zorder=6
    )

# =========================================================
# LEYENDA
# =========================================================
handles, labels = ax.get_legend_handles_labels()
unique = {}
for h, l in zip(handles, labels):
    if l not in unique:
        unique[l] = h

ax.legend(
    unique.values(), unique.keys(),
    loc="upper right",
    facecolor="black",
    edgecolor="white",
    framealpha=0.35,
    fontsize=14,
    labelcolor="white"
)

# =========================================================
# FINAL
# =========================================================
ax.set_xlim(-4.2, 4.2)
ax.set_ylim(-4.0, 4.0)
ax.set_aspect("equal")
ax.axis("off")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

np.random.seed(123)

# =========================================================
# FUNCIONES
# =========================================================
def ellipse_points(cx, cy, n, a, b, angle_deg=0):
    theta = np.random.uniform(0, 2*np.pi, n)
    r = np.sqrt(np.random.uniform(0, 1, n))
    x = a * r * np.cos(theta)
    y = b * r * np.sin(theta)

    ang = np.deg2rad(angle_deg)
    xr = x * np.cos(ang) - y * np.sin(ang)
    yr = x * np.sin(ang) + y * np.cos(ang)
    return cx + xr, cy + yr

def ring_points(cx, cy, n, a, b, angle_deg=0, rmin=0.86, rmax=1.0):
    theta = np.random.uniform(0, 2*np.pi, n)
    r = np.sqrt(np.random.uniform(rmin**2, rmax**2, n))

    x = a * r * np.cos(theta)
    y = b * r * np.sin(theta)

    ang = np.deg2rad(angle_deg)
    xr = x * np.cos(ang) - y * np.sin(ang)
    yr = x * np.sin(ang) + y * np.cos(ang)
    return cx + xr, cy + yr

def remove_inner_noise(x, y, centers, radius=1.55):
    mask = np.ones(len(x), dtype=bool)
    for cx, cy in centers:
        mask &= ((x - cx)**2 + (y - cy)**2) > radius**2
    return x[mask], y[mask]

# =========================================================
# DATOS BASE
# =========================================================
centers = [(-1.05, 1.05), (1.05, 1.05), (-1.05, -1.05), (1.05, -1.05)]
angles  = [15, -15, -10, 12]

# ruido global
n_bg = 12000
x_bg = np.random.normal(0, 2.6, n_bg)
y_bg = np.random.normal(0, 2.2, n_bg)
mask = (x_bg**2 / 4.0**2 + y_bg**2 / 3.6**2) < 1.0
x_bg, y_bg = x_bg[mask], y_bg[mask]
x_bg, y_bg = remove_inner_noise(x_bg, y_bg, centers, radius=1.55)

# market price
x_mp, y_mp = ring_points(0, 0, 900, a=3.0, b=2.7, rmin=0.88, rmax=1.0)

# componentes estructurales
components = []
for (cx, cy), ang in zip(centers, angles):
    x_ltv, y_ltv = ellipse_points(cx + 0.28, cy, 380, 0.62, 0.50, ang)
    x_pv, y_pv = ellipse_points(cx - 0.34, cy + 0.30, 360, 0.58, 0.45, ang)
    x_rs, y_rs = ellipse_points(cx - 0.34, cy - 0.30, 360, 0.58, 0.45, ang)
    x_bias, y_bias = ring_points(cx, cy, 460, 1.35, 1.05, angle_deg=ang, rmin=0.86, rmax=1.0)
    components.append({
        "ltv": (x_ltv, y_ltv),
        "pv": (x_pv, y_pv),
        "rs": (x_rs, y_rs),
        "bias": (x_bias, y_bias),
    })

# nube "aprendida" por el modelo
# inicia dispersa y luego converge al centro promedio de la estructura
target_x = []
target_y = []
for comp in components:
    for k in ["ltv", "pv", "rs"]:
        target_x.extend(comp[k][0][:80])
        target_y.extend(comp[k][1][:80])
target_x = np.array(target_x)
target_y = np.array(target_y)

n_model = 350
x_model0 = np.random.normal(0, 2.2, n_model)
y_model0 = np.random.normal(0, 1.8, n_model)

idx = np.random.choice(len(target_x), size=n_model, replace=False)
x_model1 = target_x[idx]
y_model1 = target_y[idx]

# =========================================================
# FIGURA
# =========================================================
fig, ax = plt.subplots(figsize=(10, 10), facecolor="black")
ax.set_facecolor("black")

# scatter base
noise_sc = ax.scatter([], [], s=5, c="#00cfff", alpha=0.45, edgecolors="none", label="Observed Noise")
mp_sc = ax.scatter([], [], s=20, c="#ffb000", alpha=0.75, edgecolors="none", label="Market Price")

ltv_sc = ax.scatter([], [], s=28, c="yellow", alpha=0.95, edgecolors="none", label="Long-Term Volatility")
pv_sc = ax.scatter([], [], s=28, c="red", alpha=0.95, edgecolors="none", label="Persistent Volatility")
rs_sc = ax.scatter([], [], s=28, c="green", alpha=0.95, edgecolors="none", label="Recurrent Shocks")
bias_sc = ax.scatter([], [], s=32, c="#d16cff", alpha=0.95, edgecolors="none", label="Bias")

ml_sc = ax.scatter([], [], s=26, c="white", alpha=0.9, edgecolors="none", label="ML Representation")

title = ax.text(
    0.5, 1.02, "", transform=ax.transAxes,
    ha="center", va="bottom", color="white", fontsize=16
)

ax.set_xlim(-4.2, 4.2)
ax.set_ylim(-4.0, 4.0)
ax.set_aspect("equal")
ax.axis("off")

legend = ax.legend(
    loc="upper right",
    facecolor="black",
    edgecolor="white",
    framealpha=0.35,
    fontsize=12,
    labelcolor="white"
)

# unir componentes
x_ltv_all = np.concatenate([c["ltv"][0] for c in components])
y_ltv_all = np.concatenate([c["ltv"][1] for c in components])

x_pv_all = np.concatenate([c["pv"][0] for c in components])
y_pv_all = np.concatenate([c["pv"][1] for c in components])

x_rs_all = np.concatenate([c["rs"][0] for c in components])
y_rs_all = np.concatenate([c["rs"][1] for c in components])

x_bias_all = np.concatenate([c["bias"][0] for c in components])
y_bias_all = np.concatenate([c["bias"][1] for c in components])

# =========================================================
# ANIMACIÓN
# =========================================================
frames = 100

def interpolate(a, b, t):
    return (1 - t) * a + t * b

def update(frame):
    t = frame / (frames - 1)

    # ruido se apaga gradualmente
    noise_alpha = max(0.08, 0.45 * (1 - t))
    noise_sc.set_offsets(np.column_stack([x_bg, y_bg]))
    noise_sc.set_alpha(noise_alpha)

    # market price siempre visible, un poco más claro al final
    mp_sc.set_offsets(np.column_stack([x_mp, y_mp]))
    mp_sc.set_alpha(0.45 + 0.35 * t)

    # estructura aparece gradualmente
    n_ltv = int(len(x_ltv_all) * t)
    n_pv = int(len(x_pv_all) * t)
    n_rs = int(len(x_rs_all) * t)
    n_bias = int(len(x_bias_all) * t)

    ltv_sc.set_offsets(np.column_stack([x_ltv_all[:n_ltv], y_ltv_all[:n_ltv]]) if n_ltv > 0 else np.empty((0, 2)))
    pv_sc.set_offsets(np.column_stack([x_pv_all[:n_pv], y_pv_all[:n_pv]]) if n_pv > 0 else np.empty((0, 2)))
    rs_sc.set_offsets(np.column_stack([x_rs_all[:n_rs], y_rs_all[:n_rs]]) if n_rs > 0 else np.empty((0, 2)))
    bias_sc.set_offsets(np.column_stack([x_bias_all[:n_bias], y_bias_all[:n_bias]]) if n_bias > 0 else np.empty((0, 2)))

    # nube del modelo converge desde una nube caótica a una representación estructural
    x_ml = interpolate(x_model0, x_model1, t)
    y_ml = interpolate(y_model0, y_model1, t)
    ml_sc.set_offsets(np.column_stack([x_ml, y_ml]))

    # texto dinámico
    if frame < 25:
        txt = f"Epoch {frame+1}: Raw noisy observations"
    elif frame < 55:
        txt = f"Epoch {frame+1}: Detecting persistent structure"
    elif frame < 80:
        txt = f"Epoch {frame+1}: Separating noise from volatility"
    else:
        txt = f"Epoch {frame+1}: Market representation learned"

    title.set_text(txt)

    return noise_sc, mp_sc, ltv_sc, pv_sc, rs_sc, bias_sc, ml_sc, title

ani = FuncAnimation(fig, update, frames=frames, interval=90, blit=False)

plt.show()

# Para guardar:
# ani.save("ml_noise_learning.gif", writer="pillow", fps=12)
# ani.save("ml_noise_learning.mp4", writer="ffmpeg", fps=12)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, HTMLWriter

# =========================================================
# CONFIGURACIÓN GENERAL
# =========================================================
plt.style.use("dark_background")
np.random.seed(123)

OUTPUT_FILE = "ml_market_animation_touching_noise.html"
FPS = 20
N_FRAMES = 180

# =========================================================
# FUNCIONES AUXILIARES
# =========================================================
def ellipse_points(cx, cy, n, a, b, angle_deg=0):
    """
    Genera puntos dentro de una elipse rotada.
    """
    theta = np.random.uniform(0, 2 * np.pi, n)
    r = np.sqrt(np.random.uniform(0, 1, n))
    x = a * r * np.cos(theta)
    y = b * r * np.sin(theta)

    ang = np.deg2rad(angle_deg)
    xr = x * np.cos(ang) - y * np.sin(ang)
    yr = x * np.sin(ang) + y * np.cos(ang)

    return cx + xr, cy + yr


def ring_points(cx, cy, n, a, b, angle_deg=0, rmin=0.86, rmax=1.0):
    """
    Genera puntos en un anillo elíptico rotado.
    """
    theta = np.random.uniform(0, 2 * np.pi, n)
    r = np.sqrt(np.random.uniform(rmin**2, rmax**2, n))
    x = a * r * np.cos(theta)
    y = b * r * np.sin(theta)

    ang = np.deg2rad(angle_deg)
    xr = x * np.cos(ang) - y * np.sin(ang)
    yr = x * np.sin(ang) + y * np.cos(ang)

    return cx + xr, cy + yr


def remove_inner_noise(x, y, centers, radius=1.55):
    """
    Elimina ruido del interior de las regiones de submercado.
    """
    mask = np.ones(len(x), dtype=bool)
    for cx, cy in centers:
        mask &= ((x - cx) ** 2 + (y - cy) ** 2) > radius**2
    return x[mask], y[mask]


def interpolate(a, b, t):
    return (1 - t) * a + t * b


def ease_in_out(t):
    """
    Suaviza la transición entre 0 y 1.
    """
    t = np.clip(t, 0, 1)
    return 3 * t**2 - 2 * t**3


# =========================================================
# ESTRUCTURA DEL MERCADO
# =========================================================
centers = [(-1.05, 1.05), (1.05, 1.05), (-1.05, -1.05), (1.05, -1.05)]
angles = [15, -15, -10, 12]

# =========================================================
# RUIDO GLOBAL / CAMPO OBSERVADO
# =========================================================
n_bg = 12000
x_bg = np.random.normal(0, 2.6, n_bg)
y_bg = np.random.normal(0, 2.2, n_bg)

mask_bg = (x_bg**2 / 4.0**2 + y_bg**2 / 3.6**2) < 1.0
x_bg = x_bg[mask_bg]
y_bg = y_bg[mask_bg]

# Quitar ruido del interior de los submercados
x_bg, y_bg = remove_inner_noise(x_bg, y_bg, centers, radius=1.55)

# =========================================================
# DIVISIÓN DE PUNTOS AZULES Y PUNTOS QUE PODRÁN VOLVERSE GRISES
# =========================================================
# Todos inician como azules.
# Una fracción de los puntos será "tocada" por el algoritmo y pasará a gris.
n_total = len(x_bg)

candidate_fraction = 0.50   # máximo porcentaje que puede transformarse a gris
n_candidates = int(candidate_fraction * n_total)

perm = np.random.permutation(n_total)
gray_candidate_idx = perm[:n_candidates]
blue_residual_idx = perm[n_candidates:]

# Puntos candidatos a volverse grises
x_candidates = x_bg[gray_candidate_idx]
y_candidates = y_bg[gray_candidate_idx]

# Puntos que siempre permanecerán azules
x_blue_fixed = x_bg[blue_residual_idx]
y_blue_fixed = y_bg[blue_residual_idx]

# Ordenamos candidatos para que el cambio a gris parezca una expansión
# desde el centro hacia afuera
candidate_radius = np.sqrt(x_candidates**2 + y_candidates**2)
candidate_order = np.argsort(candidate_radius)

x_candidates = x_candidates[candidate_order]
y_candidates = y_candidates[candidate_order]

# Todos los puntos azules iniciales = fijos + candidatos
x_blue_all = np.concatenate([x_blue_fixed, x_candidates])
y_blue_all = np.concatenate([y_blue_fixed, y_candidates])

# =========================================================
# PRECIO DE MERCADO GLOBAL (ANILLO NARANJA)
# =========================================================
x_mp, y_mp = ring_points(
    0, 0,
    n=900,
    a=3.0,
    b=2.7,
    angle_deg=0,
    rmin=0.88,
    rmax=1.0
)

# =========================================================
# COMPONENTES INTERNOS DE LOS SUBMERCADOS
# =========================================================
ltv_x_all, ltv_y_all = [], []
pv_x_all, pv_y_all = [], []
rs_x_all, rs_y_all = [], []

for (cx, cy), ang in zip(centers, angles):
    # Long-Term Volatility
    x_ltv, y_ltv = ellipse_points(cx + 0.28, cy, 380, 0.62, 0.50, ang)
    ltv_x_all.append(x_ltv)
    ltv_y_all.append(y_ltv)

    # Persistent Volatility
    x_pv, y_pv = ellipse_points(cx - 0.34, cy + 0.30, 360, 0.58, 0.45, ang)
    pv_x_all.append(x_pv)
    pv_y_all.append(y_pv)

    # Recurrent Shocks
    x_rs, y_rs = ellipse_points(cx - 0.34, cy - 0.30, 360, 0.58, 0.45, ang)
    rs_x_all.append(x_rs)
    rs_y_all.append(y_rs)

ltv_x_all = np.concatenate(ltv_x_all)
ltv_y_all = np.concatenate(ltv_y_all)

pv_x_all = np.concatenate(pv_x_all)
pv_y_all = np.concatenate(pv_y_all)

rs_x_all = np.concatenate(rs_x_all)
rs_y_all = np.concatenate(rs_y_all)

# =========================================================
# REPRESENTACIÓN DEL MODELO ML
# =========================================================
model_n = 420

x_model0 = np.random.normal(0, 2.0, model_n)
y_model0 = np.random.normal(0, 1.6, model_n)

pool_x = np.concatenate([ltv_x_all[:180], pv_x_all[:120], rs_x_all[:120]])
pool_y = np.concatenate([ltv_y_all[:180], pv_y_all[:120], rs_y_all[:120]])

idx = np.random.choice(len(pool_x), size=model_n, replace=True)
x_model1 = pool_x[idx]
y_model1 = pool_y[idx]

# =========================================================
# FIGURA
# =========================================================
fig, ax = plt.subplots(figsize=(10, 10))
fig.patch.set_facecolor("black")
ax.set_facecolor("black")

# Todos los azules siempre visibles
blue_sc = ax.scatter(
    [], [], s=5, c="#00cfff", alpha=0.42,
    edgecolors="none", label="Observed Information", zorder=1
)

# Puntos que el algoritmo identifica como ruido
gray_sc = ax.scatter(
    [], [], s=5, c="gray", alpha=0.0,
    edgecolors="none", label="Detected Noise", zorder=2
)

market_sc = ax.scatter(
    [], [], s=22, c="#ff9f00", alpha=0.0,
    edgecolors="none", label="Market Price Ring", zorder=6
)

ltv_sc = ax.scatter(
    [], [], s=34, c="yellow", alpha=0.97,
    edgecolors="none", label="Long-Term Volatility", zorder=4
)

pv_sc = ax.scatter(
    [], [], s=34, c="red", alpha=0.97,
    edgecolors="none", label="Persistent Volatility", zorder=4
)

rs_sc = ax.scatter(
    [], [], s=34, c="limegreen", alpha=0.97,
    edgecolors="none", label="Recurrent Shocks", zorder=4
)

ml_sc = ax.scatter(
    [], [], s=22, c="white", alpha=0.0,
    edgecolors="none", label="ML Representation", zorder=7
)

# Textos
title_text = ax.text(
    0.5, 1.03, "ML Learning Market Formation",
    transform=ax.transAxes, ha="center", va="bottom",
    color="white", fontsize=18, fontweight="bold"
)

epoch_text = ax.text(
    0.5, 0.995, "",
    transform=ax.transAxes, ha="center", va="top",
    color="white", fontsize=13
)

progress_text = ax.text(
    0.02, 0.97, "",
    transform=ax.transAxes, ha="left", va="top",
    color="white", fontsize=12
)

loss_text = ax.text(
    0.02, 0.93, "",
    transform=ax.transAxes, ha="left", va="top",
    color="white", fontsize=12
)

blue_count_text = ax.text(
    0.02, 0.89, "",
    transform=ax.transAxes, ha="left", va="top",
    color="#00cfff", fontsize=11
)

gray_count_text = ax.text(
    0.02, 0.85, "",
    transform=ax.transAxes, ha="left", va="top",
    color="lightgray", fontsize=11
)

# Barra de progreso
progress_bg, = ax.plot([], [], lw=8, alpha=0.25, solid_capstyle="round", zorder=8)
progress_fg, = ax.plot([], [], lw=8, alpha=0.95, solid_capstyle="round", zorder=9)

# Configuración del eje
ax.set_xlim(-4.2, 4.2)
ax.set_ylim(-4.0, 4.0)
ax.set_aspect("equal")
ax.axis("off")

legend = ax.legend(
    loc="upper right",
    facecolor="black",
    edgecolor="white",
    framealpha=0.35,
    fontsize=12,
    labelcolor="white"
)

# =========================================================
# ANIMACIÓN
# =========================================================
def init():
    empty = np.empty((0, 2))

    blue_sc.set_offsets(empty)
    gray_sc.set_offsets(empty)
    market_sc.set_offsets(empty)
    ltv_sc.set_offsets(empty)
    pv_sc.set_offsets(empty)
    rs_sc.set_offsets(empty)
    ml_sc.set_offsets(empty)

    epoch_text.set_text("")
    progress_text.set_text("")
    loss_text.set_text("")
    blue_count_text.set_text("")
    gray_count_text.set_text("")

    progress_bg.set_data([], [])
    progress_fg.set_data([], [])

    return (
        blue_sc, gray_sc, market_sc,
        ltv_sc, pv_sc, rs_sc, ml_sc,
        epoch_text, progress_text, loss_text,
        blue_count_text, gray_count_text,
        progress_bg, progress_fg
    )


def update(frame):
    t = frame / (N_FRAMES - 1)

    # -----------------------------------------------------
    # FASES
    # -----------------------------------------------------
    if t < 0.25:
        phase = 1
    elif t < 0.50:
        phase = 2
    elif t < 0.75:
        phase = 3
    else:
        phase = 4

    # -----------------------------------------------------
    # TEXTO
    # -----------------------------------------------------
    if phase == 1:
        stage = f"Epoch {frame + 1:03d}: Raw market field"
    elif phase == 2:
        stage = f"Epoch {frame + 1:03d}: Noise touches the field"
    elif phase == 3:
        stage = f"Epoch {frame + 1:03d}: Building submarkets"
    else:
        stage = f"Epoch {frame + 1:03d}: Market ring emerges"

    epoch_text.set_text(stage)
    progress_text.set_text(f"Training progress: {int(t * 100)}%")

    fake_loss = 1.25 * np.exp(-3.4 * t) + 0.04
    loss_text.set_text(f"Loss: {fake_loss:.3f}")

    # -----------------------------------------------------
    # BARRA DE PROGRESO
    # -----------------------------------------------------
    x0, x1 = -3.8, -1.2
    ybar = 3.55

    progress_bg.set_data([x0, x1], [ybar, ybar])
    progress_bg.set_color("gray")

    progress_fg.set_data([x0, x0 + (x1 - x0) * t], [ybar, ybar])
    progress_fg.set_color("#00cfff")

    # -----------------------------------------------------
    # CUÁNTOS PUNTOS YA FUERON "TOCADOS"
    # -----------------------------------------------------
    if phase == 1:
        k_touch = 0.0
    elif phase == 2:
        k_touch = 0.45 * ease_in_out((t - 0.25) / 0.25)
    elif phase == 3:
        k_touch = 0.45 + 0.40 * ease_in_out((t - 0.50) / 0.25)
    else:
        k_touch = 0.85 + 0.15 * ease_in_out((t - 0.75) / 0.25)

    k_touch = np.clip(k_touch, 0, 1)
    n_gray_now = int(len(x_candidates) * k_touch)

    # Los grises salen de los candidatos ya tocados
    x_gray_now = x_candidates[:n_gray_now]
    y_gray_now = y_candidates[:n_gray_now]

    # Los azules residuales son:
    # 1) los azules fijos
    # 2) los candidatos que aún no han sido tocados
    x_blue_remaining = np.concatenate([x_blue_fixed, x_candidates[n_gray_now:]])
    y_blue_remaining = np.concatenate([y_blue_fixed, y_candidates[n_gray_now:]])

    blue_count_text.set_text(f"Blue points remaining: {len(x_blue_remaining)}")
    gray_count_text.set_text(f"Gray points detected: {len(x_gray_now)}")

    # -----------------------------------------------------
    # SIEMPRE SE VEN LOS AZULES
    # -----------------------------------------------------
    blue_sc.set_offsets(np.column_stack([x_blue_remaining, y_blue_remaining]))
    gray_sc.set_offsets(np.column_stack([x_gray_now, y_gray_now]))

    # Vacíos por defecto
    empty = np.empty((0, 2))
    market_sc.set_offsets(empty)
    ltv_sc.set_offsets(empty)
    pv_sc.set_offsets(empty)
    rs_sc.set_offsets(empty)

    # -----------------------------------------------------
    # FASE 1: TODO AZUL
    # -----------------------------------------------------
    if phase == 1:
        k = ease_in_out(t / 0.25)

        blue_sc.set_alpha(0.46)
        gray_sc.set_alpha(0.0)

        # ML aún difuso
        ml_sc.set_offsets(np.column_stack([x_model0, y_model0]))
        ml_sc.set_alpha(0.08 + 0.10 * k)

    # -----------------------------------------------------
    # FASE 2: LOS GRISES TOCAN Y CAMBIAN DE COLOR
    # -----------------------------------------------------
    elif phase == 2:
        k = ease_in_out((t - 0.25) / 0.25)

        blue_sc.set_alpha(0.40)
        gray_sc.set_alpha(0.10 + 0.28 * k)

        # ML empieza a condensarse
        x_ml = interpolate(x_model0, x_model0 * 0.65, k)
        y_ml = interpolate(y_model0, y_model0 * 0.65, k)

        ml_sc.set_offsets(np.column_stack([x_ml, y_ml]))
        ml_sc.set_alpha(0.16 + 0.10 * k)

    # -----------------------------------------------------
    # FASE 3: SE FORMAN LOS SUBMERCADOS
    # -----------------------------------------------------
    elif phase == 3:
        k = ease_in_out((t - 0.50) / 0.25)

        blue_sc.set_alpha(0.32)
        gray_sc.set_alpha(0.20)

        n_ltv = max(1, int(len(ltv_x_all) * k))
        n_pv = max(1, int(len(pv_x_all) * k))
        n_rs = max(1, int(len(rs_x_all) * k))

        ltv_sc.set_offsets(np.column_stack([ltv_x_all[:n_ltv], ltv_y_all[:n_ltv]]))
        pv_sc.set_offsets(np.column_stack([pv_x_all[:n_pv], pv_y_all[:n_pv]]))
        rs_sc.set_offsets(np.column_stack([rs_x_all[:n_rs], rs_y_all[:n_rs]]))

        x_ml = interpolate(x_model0 * 0.65, x_model1, k)
        y_ml = interpolate(y_model0 * 0.65, y_model1, k)

        ml_sc.set_offsets(np.column_stack([x_ml, y_ml]))
        ml_sc.set_alpha(0.25 + 0.45 * k)

    # -----------------------------------------------------
    # FASE 4: APARECE EL ANILLO NARANJA
    # -----------------------------------------------------
    else:
        k = ease_in_out((t - 0.75) / 0.25)

        # Mantener azules residuales visibles al final
        blue_sc.set_alpha(0.34)
        gray_sc.set_alpha(0.16)

        ltv_sc.set_offsets(np.column_stack([ltv_x_all, ltv_y_all]))
        pv_sc.set_offsets(np.column_stack([pv_x_all, pv_y_all]))
        rs_sc.set_offsets(np.column_stack([rs_x_all, rs_y_all]))

        n_market = max(1, int(len(x_mp) * k))
        market_sc.set_offsets(np.column_stack([x_mp[:n_market], y_mp[:n_market]]))
        market_sc.set_alpha(0.25 + 0.55 * k)

        ml_sc.set_offsets(np.column_stack([x_model1, y_model1]))
        ml_sc.set_alpha(0.78 + 0.12 * k)

    return (
        blue_sc, gray_sc, market_sc,
        ltv_sc, pv_sc, rs_sc, ml_sc,
        epoch_text, progress_text, loss_text,
        blue_count_text, gray_count_text,
        progress_bg, progress_fg
    )


ani = FuncAnimation(
    fig,
    update,
    frames=N_FRAMES,
    init_func=init,
    interval=1000 / FPS,
    blit=False
)

# =========================================================
# GUARDAR EN HTML
# =========================================================
ani.save(OUTPUT_FILE, writer=HTMLWriter(fps=FPS))
print(f"Saved as: {OUTPUT_FILE}")

# Opcional
# plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, HTMLWriter

# =========================================================
# CONFIGURACIÓN GENERAL
# =========================================================
plt.style.use("dark_background")
np.random.seed(123)

OUTPUT_FILE = "ml_market_animation_recoloring.html"
FPS = 20
N_FRAMES = 180

# =========================================================
# FUNCIONES AUXILIARES
# =========================================================
def ellipse_points(cx, cy, n, a, b, angle_deg=0):
    """
    Genera puntos dentro de una elipse rotada.
    """
    theta = np.random.uniform(0, 2 * np.pi, n)
    r = np.sqrt(np.random.uniform(0, 1, n))
    x = a * r * np.cos(theta)
    y = b * r * np.sin(theta)

    ang = np.deg2rad(angle_deg)
    xr = x * np.cos(ang) - y * np.sin(ang)
    yr = x * np.sin(ang) + y * np.cos(ang)

    return cx + xr, cy + yr


def ring_points(cx, cy, n, a, b, angle_deg=0, rmin=0.86, rmax=1.0):
    """
    Genera puntos en un anillo elíptico rotado.
    """
    theta = np.random.uniform(0, 2 * np.pi, n)
    r = np.sqrt(np.random.uniform(rmin**2, rmax**2, n))
    x = a * r * np.cos(theta)
    y = b * r * np.sin(theta)

    ang = np.deg2rad(angle_deg)
    xr = x * np.cos(ang) - y * np.sin(ang)
    yr = x * np.sin(ang) + y * np.cos(ang)

    return cx + xr, cy + yr


def ease_in_out(t):
    t = np.clip(t, 0, 1)
    return 3 * t**2 - 2 * t**3


def interpolate(a, b, t):
    return (1 - t) * a + t * b


# =========================================================
# ESTRUCTURA DEL MERCADO
# =========================================================
centers = [(-1.05, 1.05), (1.05, 1.05), (-1.05, -1.05), (1.05, -1.05)]
angles = [15, -15, -10, 12]

# =========================================================
# 1) CAMPO INICIAL: TODO ARRANCA AZUL
# =========================================================
# Generamos un campo global azul
n_bg = 9000
x_bg = np.random.normal(0, 2.5, n_bg)
y_bg = np.random.normal(0, 2.1, n_bg)

mask_bg = (x_bg**2 / 4.0**2 + y_bg**2 / 3.6**2) < 1.0
x_bg = x_bg[mask_bg]
y_bg = y_bg[mask_bg]

# =========================================================
# 2) SUBMERCADOS: ESTOS PUNTOS TAMBIÉN NACEN AZULES
# =========================================================
ltv_x_all, ltv_y_all = [], []
pv_x_all, pv_y_all = [], []
rs_x_all, rs_y_all = [], []

for (cx, cy), ang in zip(centers, angles):
    # Long-Term Volatility (amarillo)
    x_ltv, y_ltv = ellipse_points(cx + 0.28, cy, 340, 0.62, 0.50, ang)
    ltv_x_all.append(x_ltv)
    ltv_y_all.append(y_ltv)

    # Persistent Volatility (rojo)
    x_pv, y_pv = ellipse_points(cx - 0.34, cy + 0.30, 320, 0.58, 0.45, ang)
    pv_x_all.append(x_pv)
    pv_y_all.append(y_pv)

    # Recurrent Shocks (verde)
    x_rs, y_rs = ellipse_points(cx - 0.34, cy - 0.30, 320, 0.58, 0.45, ang)
    rs_x_all.append(x_rs)
    rs_y_all.append(y_rs)

ltv_x_all = np.concatenate(ltv_x_all)
ltv_y_all = np.concatenate(ltv_y_all)

pv_x_all = np.concatenate(pv_x_all)
pv_y_all = np.concatenate(pv_y_all)

rs_x_all = np.concatenate(rs_x_all)
rs_y_all = np.concatenate(rs_y_all)

# =========================================================
# 3) ANILLO DE MERCADO
# =========================================================
x_mp, y_mp = ring_points(
    0, 0,
    n=900,
    a=3.0,
    b=2.7,
    angle_deg=0,
    rmin=0.88,
    rmax=1.0
)

# =========================================================
# 4) CONJUNTO MAESTRO: TODO EMPIEZA AZUL
# =========================================================
# Mezclamos todos los puntos que luego podrán cambiar de color
x_all = np.concatenate([x_bg, ltv_x_all, pv_x_all, rs_x_all])
y_all = np.concatenate([y_bg, ltv_y_all, pv_y_all, rs_y_all])

n_bg_final = len(x_bg)
n_ltv = len(ltv_x_all)
n_pv = len(pv_x_all)
n_rs = len(rs_x_all)

idx_bg = np.arange(0, n_bg_final)
idx_ltv = np.arange(n_bg_final, n_bg_final + n_ltv)
idx_pv = np.arange(n_bg_final + n_ltv, n_bg_final + n_ltv + n_pv)
idx_rs = np.arange(n_bg_final + n_ltv + n_pv, n_bg_final + n_ltv + n_pv + n_rs)

# =========================================================
# 5) PUNTOS DE RUIDO QUE PASARÁN A GRIS
# =========================================================
# Solo una fracción del fondo se convierte en gris
candidate_fraction = 0.55
n_gray_candidates = int(candidate_fraction * len(idx_bg))

perm_bg = np.random.permutation(idx_bg)
gray_candidates = perm_bg[:n_gray_candidates]

# Ordenamos para que el cambio empiece desde el centro y se expanda
candidate_radius = np.sqrt(x_all[gray_candidates]**2 + y_all[gray_candidates]**2)
gray_candidates = gray_candidates[np.argsort(candidate_radius)]

# =========================================================
# 6) MODELO ML
# =========================================================
model_n = 420
x_model0 = np.random.normal(0, 2.0, model_n)
y_model0 = np.random.normal(0, 1.6, model_n)

pool_x = np.concatenate([ltv_x_all[:180], pv_x_all[:120], rs_x_all[:120]])
pool_y = np.concatenate([ltv_y_all[:180], pv_y_all[:120], rs_y_all[:120]])
idx_pool = np.random.choice(len(pool_x), size=model_n, replace=True)
x_model1 = pool_x[idx_pool]
y_model1 = pool_y[idx_pool]

# =========================================================
# 7) FIGURA
# =========================================================
fig, ax = plt.subplots(figsize=(10, 10))
fig.patch.set_facecolor("black")
ax.set_facecolor("black")

# Todos los puntos que aún siguen "sin clasificar"
blue_sc = ax.scatter(
    [], [], s=5, c="#00cfff", alpha=0.42,
    edgecolors="none", label="Observed Information", zorder=1
)

# Puntos reclasificados como ruido
gray_sc = ax.scatter(
    [], [], s=5, c="gray", alpha=0.0,
    edgecolors="none", label="Detected Noise", zorder=2
)

# Submercados
ltv_sc = ax.scatter(
    [], [], s=34, c="yellow", alpha=0.95,
    edgecolors="none", label="Long-Term Volatility", zorder=4
)

pv_sc = ax.scatter(
    [], [], s=34, c="red", alpha=0.95,
    edgecolors="none", label="Persistent Volatility", zorder=4
)

rs_sc = ax.scatter(
    [], [], s=34, c="limegreen", alpha=0.95,
    edgecolors="none", label="Recurrent Shocks", zorder=4
)

# Anillo de mercado
market_sc = ax.scatter(
    [], [], s=22, c="#ff9f00", alpha=0.0,
    edgecolors="none", label="Market Price Ring", zorder=6
)

# Representación del modelo
ml_sc = ax.scatter(
    [], [], s=22, c="white", alpha=0.0,
    edgecolors="none", label="ML Representation", zorder=7
)

# Textos
title_text = ax.text(
    0.5, 1.03, "ML Learning Market Formation",
    transform=ax.transAxes, ha="center", va="bottom",
    color="white", fontsize=18, fontweight="bold"
)

epoch_text = ax.text(
    0.5, 0.995, "",
    transform=ax.transAxes, ha="center", va="top",
    color="white", fontsize=13
)

progress_text = ax.text(
    0.02, 0.97, "",
    transform=ax.transAxes, ha="left", va="top",
    color="white", fontsize=12
)

loss_text = ax.text(
    0.02, 0.93, "",
    transform=ax.transAxes, ha="left", va="top",
    color="white", fontsize=12
)

blue_count_text = ax.text(
    0.02, 0.89, "",
    transform=ax.transAxes, ha="left", va="top",
    color="#00cfff", fontsize=11
)

gray_count_text = ax.text(
    0.02, 0.85, "",
    transform=ax.transAxes, ha="left", va="top",
    color="lightgray", fontsize=11
)

# Barra de progreso
progress_bg, = ax.plot([], [], lw=8, alpha=0.25, solid_capstyle="round", zorder=8)
progress_fg, = ax.plot([], [], lw=8, alpha=0.95, solid_capstyle="round", zorder=9)

ax.set_xlim(-4.2, 4.2)
ax.set_ylim(-4.0, 4.0)
ax.set_aspect("equal")
ax.axis("off")

legend = ax.legend(
    loc="upper right",
    facecolor="black",
    edgecolor="white",
    framealpha=0.35,
    fontsize=12,
    labelcolor="white"
)

# =========================================================
# 8) INIT
# =========================================================
def init():
    empty = np.empty((0, 2))
    blue_sc.set_offsets(empty)
    gray_sc.set_offsets(empty)
    ltv_sc.set_offsets(empty)
    pv_sc.set_offsets(empty)
    rs_sc.set_offsets(empty)
    market_sc.set_offsets(empty)
    ml_sc.set_offsets(empty)

    epoch_text.set_text("")
    progress_text.set_text("")
    loss_text.set_text("")
    blue_count_text.set_text("")
    gray_count_text.set_text("")

    progress_bg.set_data([], [])
    progress_fg.set_data([], [])

    return (
        blue_sc, gray_sc, ltv_sc, pv_sc, rs_sc, market_sc, ml_sc,
        epoch_text, progress_text, loss_text,
        blue_count_text, gray_count_text,
        progress_bg, progress_fg
    )

# =========================================================
# 9) UPDATE
# =========================================================
def update(frame):
    t = frame / (N_FRAMES - 1)

    # -----------------------------------------------------
    # FASES
    # -----------------------------------------------------
    if t < 0.25:
        phase = 1
    elif t < 0.50:
        phase = 2
    elif t < 0.75:
        phase = 3
    else:
        phase = 4

    # -----------------------------------------------------
    # TEXTO
    # -----------------------------------------------------
    if phase == 1:
        stage = f"Epoch {frame + 1:03d}: Everything is observed as blue"
    elif phase == 2:
        stage = f"Epoch {frame + 1:03d}: The algorithm identifies noise"
    elif phase == 3:
        stage = f"Epoch {frame + 1:03d}: Submarkets emerge from the field"
    else:
        stage = f"Epoch {frame + 1:03d}: Market ring emerges"

    epoch_text.set_text(stage)
    progress_text.set_text(f"Training progress: {int(t * 100)}%")
    fake_loss = 1.25 * np.exp(-3.4 * t) + 0.04
    loss_text.set_text(f"Loss: {fake_loss:.3f}")

    # -----------------------------------------------------
    # BARRA
    # -----------------------------------------------------
    x0, x1 = -3.8, -1.2
    ybar = 3.55
    progress_bg.set_data([x0, x1], [ybar, ybar])
    progress_bg.set_color("gray")
    progress_fg.set_data([x0, x0 + (x1 - x0) * t], [ybar, ybar])
    progress_fg.set_color("#00cfff")

    # -----------------------------------------------------
    # RECOLOREO A GRIS
    # -----------------------------------------------------
    if phase == 1:
        k_gray = 0.0
    elif phase == 2:
        k_gray = 0.45 * ease_in_out((t - 0.25) / 0.25)
    elif phase == 3:
        k_gray = 0.45 + 0.35 * ease_in_out((t - 0.50) / 0.25)
    else:
        k_gray = 0.80 + 0.20 * ease_in_out((t - 0.75) / 0.25)

    n_gray_now = int(len(gray_candidates) * np.clip(k_gray, 0, 1))
    gray_now = gray_candidates[:n_gray_now]

    # -----------------------------------------------------
    # RECOLOREO A SUBMERCADOS
    # -----------------------------------------------------
    if phase in [1, 2]:
        k_sub = 0.0
    elif phase == 3:
        k_sub = ease_in_out((t - 0.50) / 0.25)
    else:
        k_sub = 1.0

    n_ltv_now = int(len(idx_ltv) * k_sub)
    n_pv_now = int(len(idx_pv) * k_sub)
    n_rs_now = int(len(idx_rs) * k_sub)

    ltv_now = idx_ltv[:n_ltv_now]
    pv_now = idx_pv[:n_pv_now]
    rs_now = idx_rs[:n_rs_now]

    # -----------------------------------------------------
    # AZULES = TODO MENOS LOS YA RECOLOREADOS
    # -----------------------------------------------------
    recolored = np.concatenate([gray_now, ltv_now, pv_now, rs_now]).astype(int)

    is_recolored = np.zeros(len(x_all), dtype=bool)
    is_recolored[recolored] = True

    blue_now = np.where(~is_recolored)[0]

    # -----------------------------------------------------
    # ACTUALIZAR CAPAS
    # -----------------------------------------------------
    blue_sc.set_offsets(np.column_stack([x_all[blue_now], y_all[blue_now]]))
    gray_sc.set_offsets(np.column_stack([x_all[gray_now], y_all[gray_now]]))
    ltv_sc.set_offsets(np.column_stack([x_all[ltv_now], y_all[ltv_now]]))
    pv_sc.set_offsets(np.column_stack([x_all[pv_now], y_all[pv_now]]))
    rs_sc.set_offsets(np.column_stack([x_all[rs_now], y_all[rs_now]]))

    # Transparencias
    if phase == 1:
        blue_sc.set_alpha(0.44)
        gray_sc.set_alpha(0.0)
    elif phase == 2:
        blue_sc.set_alpha(0.42)
        gray_sc.set_alpha(0.10 + 0.28 * ease_in_out((t - 0.25) / 0.25))
    elif phase == 3:
        blue_sc.set_alpha(0.36)
        gray_sc.set_alpha(0.18)
    else:
        blue_sc.set_alpha(0.34)
        gray_sc.set_alpha(0.16)

    # -----------------------------------------------------
    # ANILLO NARANJA
    # -----------------------------------------------------
    if phase == 4:
        k_ring = ease_in_out((t - 0.75) / 0.25)
        n_ring = max(1, int(len(x_mp) * k_ring))
        market_sc.set_offsets(np.column_stack([x_mp[:n_ring], y_mp[:n_ring]]))
        market_sc.set_alpha(0.25 + 0.55 * k_ring)
    else:
        market_sc.set_offsets(np.empty((0, 2)))
        market_sc.set_alpha(0.0)

    # -----------------------------------------------------
    # MODELO ML
    # -----------------------------------------------------
    if phase == 1:
        k_ml = ease_in_out(t / 0.25)
        x_ml = x_model0
        y_ml = y_model0
        ml_alpha = 0.08 + 0.08 * k_ml
    elif phase == 2:
        k_ml = ease_in_out((t - 0.25) / 0.25)
        x_ml = interpolate(x_model0, x_model0 * 0.65, k_ml)
        y_ml = interpolate(y_model0, y_model0 * 0.65, k_ml)
        ml_alpha = 0.16 + 0.08 * k_ml
    elif phase == 3:
        k_ml = ease_in_out((t - 0.50) / 0.25)
        x_ml = interpolate(x_model0 * 0.65, x_model1, k_ml)
        y_ml = interpolate(y_model0 * 0.65, y_model1, k_ml)
        ml_alpha = 0.25 + 0.45 * k_ml
    else:
        k_ml = ease_in_out((t - 0.75) / 0.25)
        x_ml = x_model1
        y_ml = y_model1
        ml_alpha = 0.78 + 0.12 * k_ml

    ml_sc.set_offsets(np.column_stack([x_ml, y_ml]))
    ml_sc.set_alpha(ml_alpha)

    blue_count_text.set_text(f"Blue points remaining: {len(blue_now)}")
    gray_count_text.set_text(f"Gray points detected: {len(gray_now)}")

    return (
        blue_sc, gray_sc, ltv_sc, pv_sc, rs_sc, market_sc, ml_sc,
        epoch_text, progress_text, loss_text,
        blue_count_text, gray_count_text,
        progress_bg, progress_fg
    )

# =========================================================
# 10) ANIMACIÓN
# =========================================================
ani = FuncAnimation(
    fig,
    update,
    frames=N_FRAMES,
    init_func=init,
    interval=1000 / FPS,
    blit=False
)

# =========================================================
# 11) GUARDAR EN HTML
# =========================================================
ani.save(OUTPUT_FILE, writer=HTMLWriter(fps=FPS))
print(f"Saved as: {OUTPUT_FILE}")

# Opcional
# plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, HTMLWriter
from matplotlib.lines import Line2D

# =========================================================
# CONFIGURACIÓN GENERAL
# =========================================================
plt.style.use("dark_background")
np.random.seed(123)

OUTPUT_FILE = "ml_market_animation_fixed_legend_and_gray.html"
FPS = 20
N_FRAMES = 190

# =========================================================
# FUNCIONES AUXILIARES
# =========================================================
def ellipse_points(cx, cy, n, a, b, angle_deg=0):
    theta = np.random.uniform(0, 2 * np.pi, n)
    r = np.sqrt(np.random.uniform(0, 1, n))
    x = a * r * np.cos(theta)
    y = b * r * np.sin(theta)

    ang = np.deg2rad(angle_deg)
    xr = x * np.cos(ang) - y * np.sin(ang)
    yr = x * np.sin(ang) + y * np.cos(ang)

    return cx + xr, cy + yr


def ring_points(cx, cy, n, a, b, angle_deg=0, rmin=0.86, rmax=1.0):
    theta = np.random.uniform(0, 2 * np.pi, n)
    r = np.sqrt(np.random.uniform(rmin**2, rmax**2, n))
    x = a * r * np.cos(theta)
    y = b * r * np.sin(theta)

    ang = np.deg2rad(angle_deg)
    xr = x * np.cos(ang) - y * np.sin(ang)
    yr = x * np.sin(ang) + y * np.cos(ang)

    return cx + xr, cy + yr


def ease_in_out(t):
    t = np.clip(t, 0, 1)
    return 3 * t**2 - 2 * t**3


def interpolate(a, b, t):
    return (1 - t) * a + t * b


def sort_by_local_radius(indices, x_all, y_all, cx=0.0, cy=0.0):
    r = np.sqrt((x_all[indices] - cx) ** 2 + (y_all[indices] - cy) ** 2)
    return indices[np.argsort(r)]


# =========================================================
# ESTRUCTURA BASE DEL MERCADO
# =========================================================
centers = [(-1.05, 1.05), (1.05, 1.05), (-1.05, -1.05), (1.05, -1.05)]
angles = [15, -15, -10, 12]

# =========================================================
# 1) CAMPO GLOBAL: TODO ARRANCA AZUL
# =========================================================
n_bg = 8500
x_bg = np.random.normal(0, 2.55, n_bg)
y_bg = np.random.normal(0, 2.15, n_bg)

mask_bg = (x_bg**2 / 4.0**2 + y_bg**2 / 3.6**2) < 1.0
x_bg = x_bg[mask_bg]
y_bg = y_bg[mask_bg]

# =========================================================
# 2) SUBMERCADOS: COMPONENTES INTERNOS + SESGO
# TODOS NACEN AZULES
# =========================================================
ltv_x_all, ltv_y_all = [], []
pv_x_all, pv_y_all = [], []
rs_x_all, rs_y_all = [], []
bias_x_all, bias_y_all = [], []

# guardamos centros aproximados para ordenar ruido interno
ltv_centers = []
pv_centers = []
rs_centers = []
bias_centers = []

for (cx, cy), ang in zip(centers, angles):
    # Long-Term Volatility
    ltv_cx, ltv_cy = cx + 0.28, cy
    x_ltv, y_ltv = ellipse_points(ltv_cx, ltv_cy, 320, 0.62, 0.50, ang)
    ltv_x_all.append(x_ltv)
    ltv_y_all.append(y_ltv)
    ltv_centers.extend([(ltv_cx, ltv_cy)] * len(x_ltv))

    # Persistent Volatility
    pv_cx, pv_cy = cx - 0.34, cy + 0.30
    x_pv, y_pv = ellipse_points(pv_cx, pv_cy, 300, 0.58, 0.45, ang)
    pv_x_all.append(x_pv)
    pv_y_all.append(y_pv)
    pv_centers.extend([(pv_cx, pv_cy)] * len(x_pv))

    # Recurrent Shocks
    rs_cx, rs_cy = cx - 0.34, cy - 0.30
    x_rs, y_rs = ellipse_points(rs_cx, rs_cy, 300, 0.58, 0.45, ang)
    rs_x_all.append(x_rs)
    rs_y_all.append(y_rs)
    rs_centers.extend([(rs_cx, rs_cy)] * len(x_rs))

    # Bias ring
    x_bias, y_bias = ring_points(
        cx, cy,
        n=380,
        a=1.32,
        b=1.02,
        angle_deg=ang,
        rmin=0.84,
        rmax=1.00
    )
    bias_x_all.append(x_bias)
    bias_y_all.append(y_bias)
    bias_centers.extend([(cx, cy)] * len(x_bias))

ltv_x_all = np.concatenate(ltv_x_all)
ltv_y_all = np.concatenate(ltv_y_all)

pv_x_all = np.concatenate(pv_x_all)
pv_y_all = np.concatenate(pv_y_all)

rs_x_all = np.concatenate(rs_x_all)
rs_y_all = np.concatenate(rs_y_all)

bias_x_all = np.concatenate(bias_x_all)
bias_y_all = np.concatenate(bias_y_all)

# =========================================================
# 3) ANILLO GLOBAL DE MERCADO
# =========================================================
x_mp, y_mp = ring_points(
    0, 0,
    n=900,
    a=3.0,
    b=2.7,
    angle_deg=0,
    rmin=0.88,
    rmax=1.0
)

# =========================================================
# 4) CONJUNTO MAESTRO: TODO EMPIEZA AZUL
# =========================================================
x_all = np.concatenate([x_bg, ltv_x_all, pv_x_all, rs_x_all, bias_x_all])
y_all = np.concatenate([y_bg, ltv_y_all, pv_y_all, rs_y_all, bias_y_all])

n_bg_final = len(x_bg)
n_ltv = len(ltv_x_all)
n_pv = len(pv_x_all)
n_rs = len(rs_x_all)
n_bias = len(bias_x_all)

idx_bg = np.arange(0, n_bg_final)
idx_ltv = np.arange(n_bg_final, n_bg_final + n_ltv)
idx_pv = np.arange(n_bg_final + n_ltv, n_bg_final + n_ltv + n_pv)
idx_rs = np.arange(n_bg_final + n_ltv + n_pv, n_bg_final + n_ltv + n_pv + n_rs)
idx_bias = np.arange(
    n_bg_final + n_ltv + n_pv + n_rs,
    n_bg_final + n_ltv + n_pv + n_rs + n_bias
)

# =========================================================
# 5) CANDIDATOS A RUIDO GRIS
# HACERLOS VISIBLES EN TODOS LOS SUBGRUPOS
# =========================================================
# Fondo global
n_gray_bg = int(0.42 * len(idx_bg))
perm_bg = np.random.permutation(idx_bg)
gray_bg_candidates = perm_bg[:n_gray_bg]
r_bg = np.sqrt(x_all[gray_bg_candidates]**2 + y_all[gray_bg_candidates]**2)
gray_bg_candidates = gray_bg_candidates[np.argsort(r_bg)]

# Más ruido interno para que se vea claramente en todos los subgrupos
n_gray_ltv = int(0.28 * len(idx_ltv))
n_gray_pv = int(0.28 * len(idx_pv))
n_gray_rs = int(0.28 * len(idx_rs))
n_gray_bias = int(0.18 * len(idx_bias))

perm_ltv = np.random.permutation(idx_ltv)
perm_pv = np.random.permutation(idx_pv)
perm_rs = np.random.permutation(idx_rs)
perm_bias = np.random.permutation(idx_bias)

gray_ltv_candidates = perm_ltv[:n_gray_ltv]
gray_pv_candidates = perm_pv[:n_gray_pv]
gray_rs_candidates = perm_rs[:n_gray_rs]
gray_bias_candidates = perm_bias[:n_gray_bias]

# ordenar localmente para que aparezcan de manera orgánica
gray_ltv_candidates = sort_by_local_radius(
    gray_ltv_candidates, x_all, y_all,
    cx=np.mean(ltv_x_all), cy=np.mean(ltv_y_all)
)
gray_pv_candidates = sort_by_local_radius(
    gray_pv_candidates, x_all, y_all,
    cx=np.mean(pv_x_all), cy=np.mean(pv_y_all)
)
gray_rs_candidates = sort_by_local_radius(
    gray_rs_candidates, x_all, y_all,
    cx=np.mean(rs_x_all), cy=np.mean(rs_y_all)
)
gray_bias_candidates = sort_by_local_radius(
    gray_bias_candidates, x_all, y_all,
    cx=0.0, cy=0.0
)

# =========================================================
# 6) MODELO ML
# =========================================================
model_n = 420
x_model0 = np.random.normal(0, 2.0, model_n)
y_model0 = np.random.normal(0, 1.6, model_n)

pool_x = np.concatenate([ltv_x_all[:180], pv_x_all[:120], rs_x_all[:120], bias_x_all[:120]])
pool_y = np.concatenate([ltv_y_all[:180], pv_y_all[:120], rs_y_all[:120], bias_y_all[:120]])
idx_pool = np.random.choice(len(pool_x), size=model_n, replace=True)
x_model1 = pool_x[idx_pool]
y_model1 = pool_y[idx_pool]

# =========================================================
# 7) FIGURA
# =========================================================
fig, ax = plt.subplots(figsize=(8, 8))
fig.patch.set_facecolor("black")
ax.set_facecolor("black")

# Quitamos "Observed Information" de la leyenda, pero mantenemos la capa azul
blue_sc = ax.scatter(
    [], [], s=5, c="#00cfff", alpha=0.42,
    edgecolors="none", zorder=1
)

# Subgrupos
ltv_sc = ax.scatter(
    [], [], s=34, c="yellow", alpha=0.95,
    edgecolors="none", zorder=4
)

pv_sc = ax.scatter(
    [], [], s=34, c="red", alpha=0.95,
    edgecolors="none", zorder=4
)

rs_sc = ax.scatter(
    [], [], s=34, c="limegreen", alpha=0.95,
    edgecolors="none", zorder=4
)

bias_sc = ax.scatter(
    [], [], s=36, c="#d16cff", alpha=0.96,
    edgecolors="none", zorder=5
)

# Grises por encima para que sí se vean en todos los subgrupos
gray_sc = ax.scatter(
    [], [], s=8, c="lightgray", alpha=0.0,
    edgecolors="none", zorder=15
)

market_sc = ax.scatter(
    [], [], s=22, c="#ff9f00", alpha=0.0,
    edgecolors="none", zorder=7
)

ml_sc = ax.scatter(
    [], [], s=22, c="white", alpha=0.0,
    edgecolors="none", zorder=8
)

title_text = ax.text(
    0.5, 1.03, "ML Learning Market Formation",
    transform=ax.transAxes, ha="center", va="bottom",
    color="white", fontsize=18, fontweight="bold"
)

epoch_text = ax.text(
    0.5, 0.995, "",
    transform=ax.transAxes, ha="center", va="top",
    color="white", fontsize=13
)

progress_text = ax.text(
    0.02, 0.97, "",
    transform=ax.transAxes, ha="left", va="top",
    color="white", fontsize=12
)

loss_text = ax.text(
    0.02, 0.93, "",
    transform=ax.transAxes, ha="left", va="top",
    color="white", fontsize=12
)

blue_count_text = ax.text(
    0.02, 0.89, "",
    transform=ax.transAxes, ha="left", va="top",
    color="#00cfff", fontsize=11
)

gray_count_text = ax.text(
    0.02, 0.85, "",
    transform=ax.transAxes, ha="left", va="top",
    color="lightgray", fontsize=11
)

progress_bg, = ax.plot([], [], lw=8, alpha=0.25, solid_capstyle="round", zorder=9)
progress_fg, = ax.plot([], [], lw=8, alpha=0.95, solid_capstyle="round", zorder=10)

ax.set_xlim(-4.2, 4.2)
ax.set_ylim(-4.0, 4.0)
ax.set_aspect("equal")
ax.axis("off")

# =========================================================
# LEYENDA MANUAL
# =========================================================
legend_elements = [
    Line2D([0], [0], marker='o', color='none', markerfacecolor='lightgray',
           markersize=7, label='Detected Noise'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='yellow',
           markersize=8, label='Long-Term Volatility'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='red',
           markersize=8, label='Persistent Volatility'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='limegreen',
           markersize=8, label='Recurrent Shocks'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='#d16cff',
           markersize=8, label='Bias'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='#ff9f00',
           markersize=7, label='Market Price Ring'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='white',
           markersize=7, label='ML Representation'),
]

ax.legend(
    handles=legend_elements,
    loc="upper left",
    bbox_to_anchor=(0.82, 0.82),   # la saca a la derecha
    borderaxespad=0.0,
    facecolor="black",
    edgecolor="white",
    framealpha=0.35,
    fontsize=12,
    labelcolor="white"
)

# =========================================================
# INIT
# =========================================================
def init():
    empty = np.empty((0, 2))

    blue_sc.set_offsets(empty)
    gray_sc.set_offsets(empty)
    ltv_sc.set_offsets(empty)
    pv_sc.set_offsets(empty)
    rs_sc.set_offsets(empty)
    bias_sc.set_offsets(empty)
    market_sc.set_offsets(empty)
    ml_sc.set_offsets(empty)

    epoch_text.set_text("")
    progress_text.set_text("")
    loss_text.set_text("")
    blue_count_text.set_text("")
    gray_count_text.set_text("")

    progress_bg.set_data([], [])
    progress_fg.set_data([], [])

    return (
        blue_sc, gray_sc, ltv_sc, pv_sc, rs_sc, bias_sc,
        market_sc, ml_sc, epoch_text, progress_text, loss_text,
        blue_count_text, gray_count_text, progress_bg, progress_fg
    )

# =========================================================
# UPDATE
# =========================================================
def update(frame):
    t = frame / (N_FRAMES - 1)

    if t < 0.25:
        phase = 1
    elif t < 0.50:
        phase = 2
    elif t < 0.78:
        phase = 3
    else:
        phase = 4

    if phase == 1:
        stage = f"Epoch {frame + 1:03d}: Everything begins in blue"
    elif phase == 2:
        stage = f"Epoch {frame + 1:03d}: The algorithm identifies noise"
    elif phase == 3:
        stage = f"Epoch {frame + 1:03d}: Submarkets and bias emerge"
    else:
        stage = f"Epoch {frame + 1:03d}: Market ring emerges"

    epoch_text.set_text(stage)
    progress_text.set_text(f"Training progress: {int(t * 100)}%")
    fake_loss = 1.25 * np.exp(-3.4 * t) + 0.04
    loss_text.set_text(f"Loss: {fake_loss:.3f}")

    # Barra
    x0, x1 = -3.8, -1.2
    ybar = 3.55
    progress_bg.set_data([x0, x1], [ybar, ybar])
    progress_bg.set_color("gray")
    progress_fg.set_data([x0, x0 + (x1 - x0) * t], [ybar, ybar])
    progress_fg.set_color("#00cfff")

    # -----------------------------------------------------
    # RUIDO GRIS
    # -----------------------------------------------------
    if phase == 1:
        k_gray_bg = 0.0
        k_gray_inner = 0.0
    elif phase == 2:
        k_gray_bg = 0.78 * ease_in_out((t - 0.25) / 0.25)
        k_gray_inner = 0.70 * ease_in_out((t - 0.25) / 0.25)
    elif phase == 3:
        k_gray_bg = 0.78 + 0.18 * ease_in_out((t - 0.50) / 0.28)
        k_gray_inner = 0.70 + 0.25 * ease_in_out((t - 0.50) / 0.28)
    else:
        k_gray_bg = 0.96
        k_gray_inner = 0.95

    n_gray_bg_now = int(len(gray_bg_candidates) * np.clip(k_gray_bg, 0, 1))
    n_gray_ltv_now = int(len(gray_ltv_candidates) * np.clip(k_gray_inner, 0, 1))
    n_gray_pv_now = int(len(gray_pv_candidates) * np.clip(k_gray_inner, 0, 1))
    n_gray_rs_now = int(len(gray_rs_candidates) * np.clip(k_gray_inner, 0, 1))
    n_gray_bias_now = int(len(gray_bias_candidates) * np.clip(0.90 * k_gray_inner, 0, 1))

    gray_now = np.concatenate([
        gray_bg_candidates[:n_gray_bg_now],
        gray_ltv_candidates[:n_gray_ltv_now],
        gray_pv_candidates[:n_gray_pv_now],
        gray_rs_candidates[:n_gray_rs_now],
        gray_bias_candidates[:n_gray_bias_now]
    ]).astype(int)

    # -----------------------------------------------------
    # SUBMERCADOS Y SESGO
    # -----------------------------------------------------
    if phase in [1, 2]:
        k_sub = 0.0
        k_bias = 0.0
    elif phase == 3:
        local_t = ease_in_out((t - 0.50) / 0.28)
        k_sub = local_t
        k_bias = 0.85 * local_t
    else:
        k_sub = 1.0
        k_bias = 1.0

    n_ltv_now = int(len(idx_ltv) * k_sub)
    n_pv_now = int(len(idx_pv) * k_sub)
    n_rs_now = int(len(idx_rs) * k_sub)
    n_bias_now = int(len(idx_bias) * k_bias)

    ltv_now = idx_ltv[:n_ltv_now]
    pv_now = idx_pv[:n_pv_now]
    rs_now = idx_rs[:n_rs_now]
    bias_now = idx_bias[:n_bias_now]

    # -----------------------------------------------------
    # AZULES = TODO MENOS LO RECOLOREADO
    # -----------------------------------------------------
    recolored = np.concatenate([gray_now, ltv_now, pv_now, rs_now, bias_now]).astype(int)

    is_recolored = np.zeros(len(x_all), dtype=bool)
    is_recolored[recolored] = True
    blue_now = np.where(~is_recolored)[0]

    # -----------------------------------------------------
    # CAPAS
    # -----------------------------------------------------
    blue_sc.set_offsets(np.column_stack([x_all[blue_now], y_all[blue_now]]))
    ltv_sc.set_offsets(np.column_stack([x_all[ltv_now], y_all[ltv_now]]))
    pv_sc.set_offsets(np.column_stack([x_all[pv_now], y_all[pv_now]]))
    rs_sc.set_offsets(np.column_stack([x_all[rs_now], y_all[rs_now]]))
    bias_sc.set_offsets(np.column_stack([x_all[bias_now], y_all[bias_now]]))
    gray_sc.set_offsets(np.column_stack([x_all[gray_now], y_all[gray_now]]))

    if phase == 1:
        blue_sc.set_alpha(0.44)
        gray_sc.set_alpha(0.0)
    elif phase == 2:
        blue_sc.set_alpha(0.42)
        gray_sc.set_alpha(0.18 + 0.20 * ease_in_out((t - 0.25) / 0.25))
    elif phase == 3:
        blue_sc.set_alpha(0.35)
        gray_sc.set_alpha(0.34)
    else:
        blue_sc.set_alpha(0.33)
        gray_sc.set_alpha(0.32)

    # -----------------------------------------------------
    # ANILLO GLOBAL
    # -----------------------------------------------------
    if phase == 4:
        k_ring = ease_in_out((t - 0.78) / 0.22)
        n_ring = max(1, int(len(x_mp) * k_ring))
        market_sc.set_offsets(np.column_stack([x_mp[:n_ring], y_mp[:n_ring]]))
        market_sc.set_alpha(0.25 + 0.55 * k_ring)
    else:
        market_sc.set_offsets(np.empty((0, 2)))
        market_sc.set_alpha(0.0)

    # -----------------------------------------------------
    # MODELO ML
    # -----------------------------------------------------
    if phase == 1:
        k_ml = ease_in_out(t / 0.25)
        x_ml = x_model0
        y_ml = y_model0
        ml_alpha = 0.08 + 0.08 * k_ml
    elif phase == 2:
        k_ml = ease_in_out((t - 0.25) / 0.25)
        x_ml = interpolate(x_model0, x_model0 * 0.65, k_ml)
        y_ml = interpolate(y_model0, y_model0 * 0.65, k_ml)
        ml_alpha = 0.16 + 0.08 * k_ml
    elif phase == 3:
        k_ml = ease_in_out((t - 0.50) / 0.28)
        x_ml = interpolate(x_model0 * 0.65, x_model1, k_ml)
        y_ml = interpolate(y_model0 * 0.65, y_model1, k_ml)
        ml_alpha = 0.25 + 0.45 * k_ml
    else:
        k_ml = ease_in_out((t - 0.78) / 0.22)
        x_ml = x_model1
        y_ml = y_model1
        ml_alpha = 0.78 + 0.12 * k_ml

    ml_sc.set_offsets(np.column_stack([x_ml, y_ml]))
    ml_sc.set_alpha(ml_alpha)

    blue_count_text.set_text(f"Blue points remaining: {len(blue_now)}")
    gray_count_text.set_text(f"Gray points detected: {len(gray_now)}")

    return (
        blue_sc, gray_sc, ltv_sc, pv_sc, rs_sc, bias_sc,
        market_sc, ml_sc, epoch_text, progress_text, loss_text,
        blue_count_text, gray_count_text, progress_bg, progress_fg
    )

# =========================================================
# ANIMACIÓN
# =========================================================
ani = FuncAnimation(
    fig,
    update,
    frames=N_FRAMES,
    init_func=init,
    interval=1000 / FPS,
    blit=False
)

# =========================================================
# GUARDAR EN HTML
# =========================================================
ani.save(OUTPUT_FILE, writer=HTMLWriter(fps=FPS))
print(f"Saved as: {OUTPUT_FILE}")

# plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, HTMLWriter
from matplotlib.lines import Line2D

# =========================================================
# CONFIGURACIÓN GENERAL
# =========================================================
plt.style.use("dark_background")
np.random.seed(123)

OUTPUT_FILE = "ml_market_animation_compact_horizontal_legend.html"
FPS = 20
N_FRAMES = 190

# =========================================================
# FUNCIONES AUXILIARES
# =========================================================
def ellipse_points(cx, cy, n, a, b, angle_deg=0):
    theta = np.random.uniform(0, 2 * np.pi, n)
    r = np.sqrt(np.random.uniform(0, 1, n))
    x = a * r * np.cos(theta)
    y = b * r * np.sin(theta)

    ang = np.deg2rad(angle_deg)
    xr = x * np.cos(ang) - y * np.sin(ang)
    yr = x * np.sin(ang) + y * np.cos(ang)

    return cx + xr, cy + yr


def ring_points(cx, cy, n, a, b, angle_deg=0, rmin=0.86, rmax=1.0):
    theta = np.random.uniform(0, 2 * np.pi, n)
    r = np.sqrt(np.random.uniform(rmin**2, rmax**2, n))
    x = a * r * np.cos(theta)
    y = b * r * np.sin(theta)

    ang = np.deg2rad(angle_deg)
    xr = x * np.cos(ang) - y * np.sin(ang)
    yr = x * np.sin(ang) + y * np.cos(ang)

    return cx + xr, cy + yr


def ease_in_out(t):
    t = np.clip(t, 0, 1)
    return 3 * t**2 - 2 * t**3


def interpolate(a, b, t):
    return (1 - t) * a + t * b


def sort_by_local_radius(indices, x_all, y_all, cx=0.0, cy=0.0):
    r = np.sqrt((x_all[indices] - cx) ** 2 + (y_all[indices] - cy) ** 2)
    return indices[np.argsort(r)]


# =========================================================
# ESTRUCTURA BASE DEL MERCADO
# =========================================================
centers = [(-1.05, 1.05), (1.05, 1.05), (-1.05, -1.05), (1.05, -1.05)]
angles = [15, -15, -10, 12]

# =========================================================
# 1) CAMPO GLOBAL: TODO ARRANCA AZUL
# =========================================================
n_bg = 8500
x_bg = np.random.normal(0, 2.55, n_bg)
y_bg = np.random.normal(0, 2.15, n_bg)

mask_bg = (x_bg**2 / 4.0**2 + y_bg**2 / 3.6**2) < 1.0
x_bg = x_bg[mask_bg]
y_bg = y_bg[mask_bg]

# =========================================================
# 2) SUBMERCADOS: COMPONENTES INTERNOS + SESGO
# =========================================================
ltv_x_all, ltv_y_all = [], []
pv_x_all, pv_y_all = [], []
rs_x_all, rs_y_all = [], []
bias_x_all, bias_y_all = [], []

ltv_centers = []
pv_centers = []
rs_centers = []
bias_centers = []

for (cx, cy), ang in zip(centers, angles):
    # Long-Term Volatility
    ltv_cx, ltv_cy = cx + 0.28, cy
    x_ltv, y_ltv = ellipse_points(ltv_cx, ltv_cy, 320, 0.62, 0.50, ang)
    ltv_x_all.append(x_ltv)
    ltv_y_all.append(y_ltv)
    ltv_centers.extend([(ltv_cx, ltv_cy)] * len(x_ltv))

    # Persistent Volatility
    pv_cx, pv_cy = cx - 0.34, cy + 0.30
    x_pv, y_pv = ellipse_points(pv_cx, pv_cy, 300, 0.58, 0.45, ang)
    pv_x_all.append(x_pv)
    pv_y_all.append(y_pv)
    pv_centers.extend([(pv_cx, pv_cy)] * len(x_pv))

    # Recurrent Shocks
    rs_cx, rs_cy = cx - 0.34, cy - 0.30
    x_rs, y_rs = ellipse_points(rs_cx, rs_cy, 300, 0.58, 0.45, ang)
    rs_x_all.append(x_rs)
    rs_y_all.append(y_rs)
    rs_centers.extend([(rs_cx, rs_cy)] * len(x_rs))

    # Bias ring
    x_bias, y_bias = ring_points(
        cx, cy,
        n=380,
        a=1.32,
        b=1.02,
        angle_deg=ang,
        rmin=0.84,
        rmax=1.00
    )
    bias_x_all.append(x_bias)
    bias_y_all.append(y_bias)
    bias_centers.extend([(cx, cy)] * len(x_bias))

ltv_x_all = np.concatenate(ltv_x_all)
ltv_y_all = np.concatenate(ltv_y_all)

pv_x_all = np.concatenate(pv_x_all)
pv_y_all = np.concatenate(pv_y_all)

rs_x_all = np.concatenate(rs_x_all)
rs_y_all = np.concatenate(rs_y_all)

bias_x_all = np.concatenate(bias_x_all)
bias_y_all = np.concatenate(bias_y_all)

# =========================================================
# 3) ANILLO GLOBAL DE MERCADO
# =========================================================
x_mp, y_mp = ring_points(
    0, 0,
    n=900,
    a=3.0,
    b=2.7,
    angle_deg=0,
    rmin=0.88,
    rmax=1.0
)

# =========================================================
# 4) CONJUNTO MAESTRO
# =========================================================
x_all = np.concatenate([x_bg, ltv_x_all, pv_x_all, rs_x_all, bias_x_all])
y_all = np.concatenate([y_bg, ltv_y_all, pv_y_all, rs_y_all, bias_y_all])

n_bg_final = len(x_bg)
n_ltv = len(ltv_x_all)
n_pv = len(pv_x_all)
n_rs = len(rs_x_all)
n_bias = len(bias_x_all)

idx_bg = np.arange(0, n_bg_final)
idx_ltv = np.arange(n_bg_final, n_bg_final + n_ltv)
idx_pv = np.arange(n_bg_final + n_ltv, n_bg_final + n_ltv + n_pv)
idx_rs = np.arange(n_bg_final + n_ltv + n_pv, n_bg_final + n_ltv + n_pv + n_rs)
idx_bias = np.arange(
    n_bg_final + n_ltv + n_pv + n_rs,
    n_bg_final + n_ltv + n_pv + n_rs + n_bias
)

# =========================================================
# 5) CANDIDATOS A RUIDO GRIS
# =========================================================
n_gray_bg = int(0.42 * len(idx_bg))
perm_bg = np.random.permutation(idx_bg)
gray_bg_candidates = perm_bg[:n_gray_bg]
r_bg = np.sqrt(x_all[gray_bg_candidates]**2 + y_all[gray_bg_candidates]**2)
gray_bg_candidates = gray_bg_candidates[np.argsort(r_bg)]

n_gray_ltv = int(0.28 * len(idx_ltv))
n_gray_pv = int(0.28 * len(idx_pv))
n_gray_rs = int(0.28 * len(idx_rs))
n_gray_bias = int(0.18 * len(idx_bias))

perm_ltv = np.random.permutation(idx_ltv)
perm_pv = np.random.permutation(idx_pv)
perm_rs = np.random.permutation(idx_rs)
perm_bias = np.random.permutation(idx_bias)

gray_ltv_candidates = perm_ltv[:n_gray_ltv]
gray_pv_candidates = perm_pv[:n_gray_pv]
gray_rs_candidates = perm_rs[:n_gray_rs]
gray_bias_candidates = perm_bias[:n_gray_bias]

gray_ltv_candidates = sort_by_local_radius(
    gray_ltv_candidates, x_all, y_all,
    cx=np.mean(ltv_x_all), cy=np.mean(ltv_y_all)
)
gray_pv_candidates = sort_by_local_radius(
    gray_pv_candidates, x_all, y_all,
    cx=np.mean(pv_x_all), cy=np.mean(pv_y_all)
)
gray_rs_candidates = sort_by_local_radius(
    gray_rs_candidates, x_all, y_all,
    cx=np.mean(rs_x_all), cy=np.mean(rs_y_all)
)
gray_bias_candidates = sort_by_local_radius(
    gray_bias_candidates, x_all, y_all,
    cx=0.0, cy=0.0
)

# =========================================================
# 6) MODELO ML
# =========================================================
model_n = 420
x_model0 = np.random.normal(0, 2.0, model_n)
y_model0 = np.random.normal(0, 1.6, model_n)

pool_x = np.concatenate([ltv_x_all[:180], pv_x_all[:120], rs_x_all[:120], bias_x_all[:120]])
pool_y = np.concatenate([ltv_y_all[:180], pv_y_all[:120], rs_y_all[:120], bias_y_all[:120]])
idx_pool = np.random.choice(len(pool_x), size=model_n, replace=True)
x_model1 = pool_x[idx_pool]
y_model1 = pool_y[idx_pool]

# =========================================================
# 7) FIGURA COMPACTA PARA HTML
# =========================================================
fig, ax = plt.subplots(figsize=(12, 7))
fig.patch.set_facecolor("black")
ax.set_facecolor("black")

blue_sc = ax.scatter(
    [], [], s=5, c="#00cfff", alpha=0.42,
    edgecolors="none", zorder=1
)

ltv_sc = ax.scatter(
    [], [], s=34, c="yellow", alpha=0.95,
    edgecolors="none", zorder=4
)

pv_sc = ax.scatter(
    [], [], s=34, c="red", alpha=0.95,
    edgecolors="none", zorder=4
)

rs_sc = ax.scatter(
    [], [], s=34, c="limegreen", alpha=0.95,
    edgecolors="none", zorder=4
)

bias_sc = ax.scatter(
    [], [], s=36, c="#d16cff", alpha=0.96,
    edgecolors="none", zorder=5
)

gray_sc = ax.scatter(
    [], [], s=8, c="lightgray", alpha=0.0,
    edgecolors="none", zorder=15
)

market_sc = ax.scatter(
    [], [], s=22, c="#ff9f00", alpha=0.0,
    edgecolors="none", zorder=7
)

ml_sc = ax.scatter(
    [], [], s=22, c="white", alpha=0.0,
    edgecolors="none", zorder=8
)

# =========================================================
# TEXTOS MÁS COMPACTOS
# =========================================================
title_text = ax.text(
    0.5, 0.985, "ML Learning Market Formation",
    transform=ax.transAxes, ha="center", va="top",
    color="white", fontsize=17, fontweight="bold"
)

epoch_text = ax.text(
    0.5, 0.945, "",
    transform=ax.transAxes, ha="center", va="top",
    color="white", fontsize=12
)

progress_text = ax.text(
    0.02, 0.95, "",
    transform=ax.transAxes, ha="left", va="top",
    color="white", fontsize=11
)

loss_text = ax.text(
    0.02, 0.91, "",
    transform=ax.transAxes, ha="left", va="top",
    color="white", fontsize=11
)

blue_count_text = ax.text(
    0.02, 0.87, "",
    transform=ax.transAxes, ha="left", va="top",
    color="#00cfff", fontsize=10.5
)

gray_count_text = ax.text(
    0.02, 0.83, "",
    transform=ax.transAxes, ha="left", va="top",
    color="lightgray", fontsize=10.5
)

progress_bg, = ax.plot([], [], lw=7, alpha=0.25, solid_capstyle="round", zorder=9)
progress_fg, = ax.plot([], [], lw=7, alpha=0.95, solid_capstyle="round", zorder=10)

# =========================================================
# MÁS COMPACTO EN VERTICAL
# =========================================================
ax.set_xlim(-4.2, 4.2)
ax.set_ylim(-3.15, 3.15)
ax.set_aspect("equal")
ax.axis("off")

# =========================================================
# LEYENDA HORIZONTAL ABAJO
# =========================================================
legend_elements = [
    Line2D([0], [0], marker='o', color='none', markerfacecolor='lightgray',
           markersize=7, label='Detected Noise'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='yellow',
           markersize=8, label='Long-Term Volatility'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='red',
           markersize=8, label='Persistent Volatility'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='limegreen',
           markersize=8, label='Recurrent Shocks'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='#d16cff',
           markersize=8, label='Bias'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='#ff9f00',
           markersize=7, label='Market Price Ring'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='white',
           markersize=7, label='ML Representation'),
]

legend = ax.legend(
    handles=legend_elements,
    loc="upper center",
    bbox_to_anchor=(0.5, -0.06),
    ncol=4,
    facecolor="black",
    edgecolor="white",
    framealpha=0.25,
    fontsize=10.5,
    labelcolor="white",
    handletextpad=0.5,
    columnspacing=1.4,
    borderpad=0.6
)

# =========================================================
# INIT
# =========================================================
def init():
    empty = np.empty((0, 2))

    blue_sc.set_offsets(empty)
    gray_sc.set_offsets(empty)
    ltv_sc.set_offsets(empty)
    pv_sc.set_offsets(empty)
    rs_sc.set_offsets(empty)
    bias_sc.set_offsets(empty)
    market_sc.set_offsets(empty)
    ml_sc.set_offsets(empty)

    epoch_text.set_text("")
    progress_text.set_text("")
    loss_text.set_text("")
    blue_count_text.set_text("")
    gray_count_text.set_text("")

    progress_bg.set_data([], [])
    progress_fg.set_data([], [])

    return (
        blue_sc, gray_sc, ltv_sc, pv_sc, rs_sc, bias_sc,
        market_sc, ml_sc, epoch_text, progress_text, loss_text,
        blue_count_text, gray_count_text, progress_bg, progress_fg
    )

# =========================================================
# UPDATE
# =========================================================
def update(frame):
    t = frame / (N_FRAMES - 1)

    if t < 0.25:
        phase = 1
    elif t < 0.50:
        phase = 2
    elif t < 0.78:
        phase = 3
    else:
        phase = 4

    if phase == 1:
        stage = f"Epoch {frame + 1:03d}: Everything begins in blue"
    elif phase == 2:
        stage = f"Epoch {frame + 1:03d}: The algorithm identifies noise"
    elif phase == 3:
        stage = f"Epoch {frame + 1:03d}: Submarkets and bias emerge"
    else:
        stage = f"Epoch {frame + 1:03d}: Market ring emerges"

    epoch_text.set_text(stage)
    progress_text.set_text(f"Training progress: {int(t * 100)}%")
    fake_loss = 1.25 * np.exp(-3.4 * t) + 0.04
    loss_text.set_text(f"Loss: {fake_loss:.3f}")

    # Barra de progreso más baja y compacta
    x0, x1 = -3.8, -1.6
    ybar = 2.5
    progress_bg.set_data([x0, x1], [ybar, ybar])
    progress_bg.set_color("gray")
    progress_fg.set_data([x0, x0 + (x1 - x0) * t], [ybar, ybar])
    progress_fg.set_color("#00cfff")

    # -----------------------------------------------------
    # RUIDO GRIS
    # -----------------------------------------------------
    if phase == 1:
        k_gray_bg = 0.0
        k_gray_inner = 0.0
    elif phase == 2:
        k_gray_bg = 0.78 * ease_in_out((t - 0.25) / 0.25)
        k_gray_inner = 0.70 * ease_in_out((t - 0.25) / 0.25)
    elif phase == 3:
        k_gray_bg = 0.78 + 0.18 * ease_in_out((t - 0.50) / 0.28)
        k_gray_inner = 0.70 + 0.25 * ease_in_out((t - 0.50) / 0.28)
    else:
        k_gray_bg = 0.96
        k_gray_inner = 0.95

    n_gray_bg_now = int(len(gray_bg_candidates) * np.clip(k_gray_bg, 0, 1))
    n_gray_ltv_now = int(len(gray_ltv_candidates) * np.clip(k_gray_inner, 0, 1))
    n_gray_pv_now = int(len(gray_pv_candidates) * np.clip(k_gray_inner, 0, 1))
    n_gray_rs_now = int(len(gray_rs_candidates) * np.clip(k_gray_inner, 0, 1))
    n_gray_bias_now = int(len(gray_bias_candidates) * np.clip(0.90 * k_gray_inner, 0, 1))

    gray_now = np.concatenate([
        gray_bg_candidates[:n_gray_bg_now],
        gray_ltv_candidates[:n_gray_ltv_now],
        gray_pv_candidates[:n_gray_pv_now],
        gray_rs_candidates[:n_gray_rs_now],
        gray_bias_candidates[:n_gray_bias_now]
    ]).astype(int)

    # -----------------------------------------------------
    # SUBMERCADOS Y SESGO
    # -----------------------------------------------------
    if phase in [1, 2]:
        k_sub = 0.0
        k_bias = 0.0
    elif phase == 3:
        local_t = ease_in_out((t - 0.50) / 0.28)
        k_sub = local_t
        k_bias = 0.85 * local_t
    else:
        k_sub = 1.0
        k_bias = 1.0

    n_ltv_now = int(len(idx_ltv) * k_sub)
    n_pv_now = int(len(idx_pv) * k_sub)
    n_rs_now = int(len(idx_rs) * k_sub)
    n_bias_now = int(len(idx_bias) * k_bias)

    ltv_now = idx_ltv[:n_ltv_now]
    pv_now = idx_pv[:n_pv_now]
    rs_now = idx_rs[:n_rs_now]
    bias_now = idx_bias[:n_bias_now]

    # -----------------------------------------------------
    # AZULES = TODO MENOS LO RECOLOREADO
    # -----------------------------------------------------
    recolored = np.concatenate([gray_now, ltv_now, pv_now, rs_now, bias_now]).astype(int)

    is_recolored = np.zeros(len(x_all), dtype=bool)
    is_recolored[recolored] = True
    blue_now = np.where(~is_recolored)[0]

    # -----------------------------------------------------
    # CAPAS
    # -----------------------------------------------------
    blue_sc.set_offsets(np.column_stack([x_all[blue_now], y_all[blue_now]]))
    ltv_sc.set_offsets(np.column_stack([x_all[ltv_now], y_all[ltv_now]]))
    pv_sc.set_offsets(np.column_stack([x_all[pv_now], y_all[pv_now]]))
    rs_sc.set_offsets(np.column_stack([x_all[rs_now], y_all[rs_now]]))
    bias_sc.set_offsets(np.column_stack([x_all[bias_now], y_all[bias_now]]))
    gray_sc.set_offsets(np.column_stack([x_all[gray_now], y_all[gray_now]]))

    if phase == 1:
        blue_sc.set_alpha(0.44)
        gray_sc.set_alpha(0.0)
    elif phase == 2:
        blue_sc.set_alpha(0.42)
        gray_sc.set_alpha(0.18 + 0.20 * ease_in_out((t - 0.25) / 0.25))
    elif phase == 3:
        blue_sc.set_alpha(0.35)
        gray_sc.set_alpha(0.34)
    else:
        blue_sc.set_alpha(0.33)
        gray_sc.set_alpha(0.32)

    # -----------------------------------------------------
    # ANILLO GLOBAL
    # -----------------------------------------------------
    if phase == 4:
        k_ring = ease_in_out((t - 0.78) / 0.22)
        n_ring = max(1, int(len(x_mp) * k_ring))
        market_sc.set_offsets(np.column_stack([x_mp[:n_ring], y_mp[:n_ring]]))
        market_sc.set_alpha(0.25 + 0.55 * k_ring)
    else:
        market_sc.set_offsets(np.empty((0, 2)))
        market_sc.set_alpha(0.0)

    # -----------------------------------------------------
    # MODELO ML
    # -----------------------------------------------------
    if phase == 1:
        k_ml = ease_in_out(t / 0.25)
        x_ml = x_model0
        y_ml = y_model0
        ml_alpha = 0.08 + 0.08 * k_ml
    elif phase == 2:
        k_ml = ease_in_out((t - 0.25) / 0.25)
        x_ml = interpolate(x_model0, x_model0 * 0.65, k_ml)
        y_ml = interpolate(y_model0, y_model0 * 0.65, k_ml)
        ml_alpha = 0.16 + 0.08 * k_ml
    elif phase == 3:
        k_ml = ease_in_out((t - 0.50) / 0.28)
        x_ml = interpolate(x_model0 * 0.65, x_model1, k_ml)
        y_ml = interpolate(y_model0 * 0.65, y_model1, k_ml)
        ml_alpha = 0.25 + 0.45 * k_ml
    else:
        k_ml = ease_in_out((t - 0.78) / 0.22)
        x_ml = x_model1
        y_ml = y_model1
        ml_alpha = 0.78 + 0.12 * k_ml

    ml_sc.set_offsets(np.column_stack([x_ml, y_ml]))
    ml_sc.set_alpha(ml_alpha)

    blue_count_text.set_text(f"Blue points remaining: {len(blue_now)}")
    gray_count_text.set_text(f"Gray points detected: {len(gray_now)}")

    return (
        blue_sc, gray_sc, ltv_sc, pv_sc, rs_sc, bias_sc,
        market_sc, ml_sc, epoch_text, progress_text, loss_text,
        blue_count_text, gray_count_text, progress_bg, progress_fg
    )

# =========================================================
# ANIMACIÓN
# =========================================================
ani = FuncAnimation(
    fig,
    update,
    frames=N_FRAMES,
    init_func=init,
    interval=1000 / FPS,
    blit=False
)

# =========================================================
# AJUSTE DE MÁRGENES PARA HTML
# =========================================================
fig.subplots_adjust(
    left=0.03,
    right=0.97,
    top=0.96,
    bottom=0.16
)

# =========================================================
# GUARDAR EN HTML
# =========================================================
ani.save(OUTPUT_FILE, writer=HTMLWriter(fps=FPS))
print(f"Saved as: {OUTPUT_FILE}")

# plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, HTMLWriter
from matplotlib.lines import Line2D

# =========================================================
# CONFIGURACIÓN GENERAL
# =========================================================
plt.style.use("dark_background")
np.random.seed(123)

OUTPUT_FILE = "ml_market_animation_clean_layout.html"
FPS = 20
N_FRAMES = 190

# =========================================================
# FUNCIONES AUXILIARES
# =========================================================
def ellipse_points(cx, cy, n, a, b, angle_deg=0):
    theta = np.random.uniform(0, 2 * np.pi, n)
    r = np.sqrt(np.random.uniform(0, 1, n))
    x = a * r * np.cos(theta)
    y = b * r * np.sin(theta)

    ang = np.deg2rad(angle_deg)
    xr = x * np.cos(ang) - y * np.sin(ang)
    yr = x * np.sin(ang) + y * np.cos(ang)

    return cx + xr, cy + yr


def ring_points(cx, cy, n, a, b, angle_deg=0, rmin=0.86, rmax=1.0):
    theta = np.random.uniform(0, 2 * np.pi, n)
    r = np.sqrt(np.random.uniform(rmin**2, rmax**2, n))
    x = a * r * np.cos(theta)
    y = b * r * np.sin(theta)

    ang = np.deg2rad(angle_deg)
    xr = x * np.cos(ang) - y * np.sin(ang)
    yr = x * np.sin(ang) + y * np.cos(ang)

    return cx + xr, cy + yr


def ease_in_out(t):
    t = np.clip(t, 0, 1)
    return 3 * t**2 - 2 * t**3


def interpolate(a, b, t):
    return (1 - t) * a + t * b


def sort_by_local_radius(indices, x_all, y_all, cx=0.0, cy=0.0):
    r = np.sqrt((x_all[indices] - cx) ** 2 + (y_all[indices] - cy) ** 2)
    return indices[np.argsort(r)]


# =========================================================
# ESTRUCTURA BASE DEL MERCADO
# =========================================================
centers = [(-1.05, 1.05), (1.05, 1.05), (-1.05, -1.05), (1.05, -1.05)]
angles = [15, -15, -10, 12]

# =========================================================
# 1) CAMPO GLOBAL: TODO ARRANCA AZUL
# =========================================================
n_bg = 8500
x_bg = np.random.normal(0, 2.55, n_bg)
y_bg = np.random.normal(0, 2.15, n_bg)

mask_bg = (x_bg**2 / 4.0**2 + y_bg**2 / 3.6**2) < 1.0
x_bg = x_bg[mask_bg]
y_bg = y_bg[mask_bg]

# =========================================================
# 2) SUBMERCADOS
# =========================================================
ltv_x_all, ltv_y_all = [], []
pv_x_all, pv_y_all = [], []
rs_x_all, rs_y_all = [], []
bias_x_all, bias_y_all = [], []

for (cx, cy), ang in zip(centers, angles):
    ltv_cx, ltv_cy = cx + 0.28, cy
    x_ltv, y_ltv = ellipse_points(ltv_cx, ltv_cy, 320, 0.62, 0.50, ang)
    ltv_x_all.append(x_ltv)
    ltv_y_all.append(y_ltv)

    pv_cx, pv_cy = cx - 0.34, cy + 0.30
    x_pv, y_pv = ellipse_points(pv_cx, pv_cy, 300, 0.58, 0.45, ang)
    pv_x_all.append(x_pv)
    pv_y_all.append(y_pv)

    rs_cx, rs_cy = cx - 0.34, cy - 0.30
    x_rs, y_rs = ellipse_points(rs_cx, rs_cy, 300, 0.58, 0.45, ang)
    rs_x_all.append(x_rs)
    rs_y_all.append(y_rs)

    x_bias, y_bias = ring_points(
        cx, cy,
        n=380,
        a=1.32,
        b=1.02,
        angle_deg=ang,
        rmin=0.84,
        rmax=1.00
    )
    bias_x_all.append(x_bias)
    bias_y_all.append(y_bias)

ltv_x_all = np.concatenate(ltv_x_all)
ltv_y_all = np.concatenate(ltv_y_all)
pv_x_all = np.concatenate(pv_x_all)
pv_y_all = np.concatenate(pv_y_all)
rs_x_all = np.concatenate(rs_x_all)
rs_y_all = np.concatenate(rs_y_all)
bias_x_all = np.concatenate(bias_x_all)
bias_y_all = np.concatenate(bias_y_all)

# =========================================================
# 3) ANILLO GLOBAL
# =========================================================
x_mp, y_mp = ring_points(
    0, 0,
    n=900,
    a=3.0,
    b=2.7,
    angle_deg=0,
    rmin=0.88,
    rmax=1.0
)

# =========================================================
# 4) CONJUNTO MAESTRO
# =========================================================
x_all = np.concatenate([x_bg, ltv_x_all, pv_x_all, rs_x_all, bias_x_all])
y_all = np.concatenate([y_bg, ltv_y_all, pv_y_all, rs_y_all, bias_y_all])

n_bg_final = len(x_bg)
n_ltv = len(ltv_x_all)
n_pv = len(pv_x_all)
n_rs = len(rs_x_all)
n_bias = len(bias_x_all)

idx_bg = np.arange(0, n_bg_final)
idx_ltv = np.arange(n_bg_final, n_bg_final + n_ltv)
idx_pv = np.arange(n_bg_final + n_ltv, n_bg_final + n_ltv + n_pv)
idx_rs = np.arange(n_bg_final + n_ltv + n_pv, n_bg_final + n_ltv + n_pv + n_rs)
idx_bias = np.arange(
    n_bg_final + n_ltv + n_pv + n_rs,
    n_bg_final + n_ltv + n_pv + n_rs + n_bias
)

# =========================================================
# 5) CANDIDATOS A RUIDO GRIS
# =========================================================
n_gray_bg = int(0.42 * len(idx_bg))
perm_bg = np.random.permutation(idx_bg)
gray_bg_candidates = perm_bg[:n_gray_bg]
r_bg = np.sqrt(x_all[gray_bg_candidates]**2 + y_all[gray_bg_candidates]**2)
gray_bg_candidates = gray_bg_candidates[np.argsort(r_bg)]

n_gray_ltv = int(0.28 * len(idx_ltv))
n_gray_pv = int(0.28 * len(idx_pv))
n_gray_rs = int(0.28 * len(idx_rs))
n_gray_bias = int(0.18 * len(idx_bias))

perm_ltv = np.random.permutation(idx_ltv)
perm_pv = np.random.permutation(idx_pv)
perm_rs = np.random.permutation(idx_rs)
perm_bias = np.random.permutation(idx_bias)

gray_ltv_candidates = sort_by_local_radius(
    perm_ltv[:n_gray_ltv], x_all, y_all,
    cx=np.mean(ltv_x_all), cy=np.mean(ltv_y_all)
)
gray_pv_candidates = sort_by_local_radius(
    perm_pv[:n_gray_pv], x_all, y_all,
    cx=np.mean(pv_x_all), cy=np.mean(pv_y_all)
)
gray_rs_candidates = sort_by_local_radius(
    perm_rs[:n_gray_rs], x_all, y_all,
    cx=np.mean(rs_x_all), cy=np.mean(rs_y_all)
)
gray_bias_candidates = sort_by_local_radius(
    perm_bias[:n_gray_bias], x_all, y_all,
    cx=0.0, cy=0.0
)

# =========================================================
# 6) MODELO ML
# =========================================================
model_n = 420
x_model0 = np.random.normal(0, 2.0, model_n)
y_model0 = np.random.normal(0, 1.6, model_n)

pool_x = np.concatenate([ltv_x_all[:180], pv_x_all[:120], rs_x_all[:120], bias_x_all[:120]])
pool_y = np.concatenate([ltv_y_all[:180], pv_y_all[:120], rs_y_all[:120], bias_y_all[:120]])
idx_pool = np.random.choice(len(pool_x), size=model_n, replace=True)
x_model1 = pool_x[idx_pool]
y_model1 = pool_y[idx_pool]

# =========================================================
# 7) FIGURA
# =========================================================
fig, ax = plt.subplots(figsize=(13, 7))
fig.patch.set_facecolor("black")
ax.set_facecolor("black")

blue_sc = ax.scatter([], [], s=5, c="#00cfff", alpha=0.42, edgecolors="none", zorder=1)
ltv_sc = ax.scatter([], [], s=34, c="yellow", alpha=0.95, edgecolors="none", zorder=4)
pv_sc = ax.scatter([], [], s=34, c="red", alpha=0.95, edgecolors="none", zorder=4)
rs_sc = ax.scatter([], [], s=34, c="limegreen", alpha=0.95, edgecolors="none", zorder=4)
bias_sc = ax.scatter([], [], s=36, c="#d16cff", alpha=0.96, edgecolors="none", zorder=5)
gray_sc = ax.scatter([], [], s=8, c="lightgray", alpha=0.0, edgecolors="none", zorder=15)
market_sc = ax.scatter([], [], s=22, c="#ff9f00", alpha=0.0, edgecolors="none", zorder=7)
ml_sc = ax.scatter([], [], s=22, c="white", alpha=0.0, edgecolors="none", zorder=8)

# =========================================================
# TEXTOS BIEN SEPARADOS
# =========================================================
title_text = ax.text(
    0.5, 0.975, "ML Learning Market Formation",
    transform=ax.transAxes, ha="center", va="top",
    color="white", fontsize=20, fontweight="bold"
)

epoch_text = ax.text(
    0.5, 0.905, "",
    transform=ax.transAxes, ha="center", va="top",
    color="white", fontsize=15
)

# bloque informativo a la izquierda y más abajo
progress_text = ax.text(
    0.03, 0.89, "",
    transform=ax.transAxes, ha="left", va="center",
    color="white", fontsize=15
)

loss_text = ax.text(
    0.03, 0.82, "",
    transform=ax.transAxes, ha="left", va="center",
    color="white", fontsize=15
)

blue_count_text = ax.text(
    0.03, 0.75, "",
    transform=ax.transAxes, ha="left", va="center",
    color="#00cfff", fontsize=15
)

gray_count_text = ax.text(
    0.03, 0.69, "",
    transform=ax.transAxes, ha="left", va="center",
    color="lightgray", fontsize=15
)

# barra en coordenadas del eje, no del dato
progress_bg, = ax.plot([], [], lw=11, alpha=0.28, solid_capstyle="round", zorder=20)
progress_fg, = ax.plot([], [], lw=11, alpha=0.95, solid_capstyle="round", zorder=21)
progress_bg.set_transform(ax.transAxes)
progress_fg.set_transform(ax.transAxes)

# =========================================================
# LÍMITES
# =========================================================
ax.set_xlim(-4.2, 4.2)
ax.set_ylim(-3.3, 2.9)   # baja visualmente la nube de puntos
ax.set_aspect("equal")
ax.axis("off")

# =========================================================
# LEYENDA HORIZONTAL ABAJO
# =========================================================
legend_elements = [
    Line2D([0], [0], marker='o', color='none', markerfacecolor='lightgray', markersize=7, label='Detected Noise'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='yellow', markersize=8, label='Long-Term Volatility'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='red', markersize=8, label='Persistent Volatility'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='limegreen', markersize=8, label='Recurrent Shocks'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='#d16cff', markersize=8, label='Bias'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='#ff9f00', markersize=7, label='Market Price Ring'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='white', markersize=7, label='ML Representation'),
]

ax.legend(
    handles=legend_elements,
    loc="upper center",
    bbox_to_anchor=(0.5, -0.055),
    ncol=4,
    facecolor="black",
    edgecolor="white",
    framealpha=0.22,
    fontsize=10.5,
    labelcolor="white",
    handletextpad=0.45,
    columnspacing=1.25,
    borderpad=0.55
)

# =========================================================
# INIT
# =========================================================
def init():
    empty = np.empty((0, 2))

    blue_sc.set_offsets(empty)
    gray_sc.set_offsets(empty)
    ltv_sc.set_offsets(empty)
    pv_sc.set_offsets(empty)
    rs_sc.set_offsets(empty)
    bias_sc.set_offsets(empty)
    market_sc.set_offsets(empty)
    ml_sc.set_offsets(empty)

    epoch_text.set_text("")
    progress_text.set_text("")
    loss_text.set_text("")
    blue_count_text.set_text("")
    gray_count_text.set_text("")

    progress_bg.set_data([], [])
    progress_fg.set_data([], [])

    return (
        blue_sc, gray_sc, ltv_sc, pv_sc, rs_sc, bias_sc,
        market_sc, ml_sc, epoch_text, progress_text, loss_text,
        blue_count_text, gray_count_text, progress_bg, progress_fg
    )

# =========================================================
# UPDATE
# =========================================================
def update(frame):
    t = frame / (N_FRAMES - 1)

    if t < 0.25:
        phase = 1
    elif t < 0.50:
        phase = 2
    elif t < 0.78:
        phase = 3
    else:
        phase = 4

    if phase == 1:
        stage = f"Epoch {frame + 1:03d}: Everything begins in blue"
    elif phase == 2:
        stage = f"Epoch {frame + 1:03d}: The algorithm identifies noise"
    elif phase == 3:
        stage = f"Epoch {frame + 1:03d}: Submarkets and bias emerge"
    else:
        stage = f"Epoch {frame + 1:03d}: Market ring emerges"

    epoch_text.set_text(stage)
    progress_text.set_text(f"Training progress: {int(t * 100)}%")

    fake_loss = 1.25 * np.exp(-3.4 * t) + 0.04
    loss_text.set_text(f"Loss: {fake_loss:.3f}")

    # barra debajo de training progress
    x0, x1 = 0.03, 0.34
    ybar = 0.845
    progress_bg.set_data([x0, x1], [ybar, ybar])
    progress_fg.set_data([x0, x0 + (x1 - x0) * t], [ybar, ybar])
    progress_bg.set_color("#4a4a4a")
    progress_fg.set_color("#00cfff")

    # ruido gris
    if phase == 1:
        k_gray_bg = 0.0
        k_gray_inner = 0.0
    elif phase == 2:
        k_gray_bg = 0.78 * ease_in_out((t - 0.25) / 0.25)
        k_gray_inner = 0.70 * ease_in_out((t - 0.25) / 0.25)
    elif phase == 3:
        k_gray_bg = 0.78 + 0.18 * ease_in_out((t - 0.50) / 0.28)
        k_gray_inner = 0.70 + 0.25 * ease_in_out((t - 0.50) / 0.28)
    else:
        k_gray_bg = 0.96
        k_gray_inner = 0.95

    n_gray_bg_now = int(len(gray_bg_candidates) * np.clip(k_gray_bg, 0, 1))
    n_gray_ltv_now = int(len(gray_ltv_candidates) * np.clip(k_gray_inner, 0, 1))
    n_gray_pv_now = int(len(gray_pv_candidates) * np.clip(k_gray_inner, 0, 1))
    n_gray_rs_now = int(len(gray_rs_candidates) * np.clip(k_gray_inner, 0, 1))
    n_gray_bias_now = int(len(gray_bias_candidates) * np.clip(0.90 * k_gray_inner, 0, 1))

    gray_now = np.concatenate([
        gray_bg_candidates[:n_gray_bg_now],
        gray_ltv_candidates[:n_gray_ltv_now],
        gray_pv_candidates[:n_gray_pv_now],
        gray_rs_candidates[:n_gray_rs_now],
        gray_bias_candidates[:n_gray_bias_now]
    ]).astype(int)

    # submercados
    if phase in [1, 2]:
        k_sub = 0.0
        k_bias = 0.0
    elif phase == 3:
        local_t = ease_in_out((t - 0.50) / 0.28)
        k_sub = local_t
        k_bias = 0.85 * local_t
    else:
        k_sub = 1.0
        k_bias = 1.0

    n_ltv_now = int(len(idx_ltv) * k_sub)
    n_pv_now = int(len(idx_pv) * k_sub)
    n_rs_now = int(len(idx_rs) * k_sub)
    n_bias_now = int(len(idx_bias) * k_bias)

    ltv_now = idx_ltv[:n_ltv_now]
    pv_now = idx_pv[:n_pv_now]
    rs_now = idx_rs[:n_rs_now]
    bias_now = idx_bias[:n_bias_now]

    recolored = np.concatenate([gray_now, ltv_now, pv_now, rs_now, bias_now]).astype(int)

    is_recolored = np.zeros(len(x_all), dtype=bool)
    is_recolored[recolored] = True
    blue_now = np.where(~is_recolored)[0]

    blue_sc.set_offsets(np.column_stack([x_all[blue_now], y_all[blue_now]]))
    ltv_sc.set_offsets(np.column_stack([x_all[ltv_now], y_all[ltv_now]]))
    pv_sc.set_offsets(np.column_stack([x_all[pv_now], y_all[pv_now]]))
    rs_sc.set_offsets(np.column_stack([x_all[rs_now], y_all[rs_now]]))
    bias_sc.set_offsets(np.column_stack([x_all[bias_now], y_all[bias_now]]))
    gray_sc.set_offsets(np.column_stack([x_all[gray_now], y_all[gray_now]]))

    if phase == 1:
        blue_sc.set_alpha(0.44)
        gray_sc.set_alpha(0.0)
    elif phase == 2:
        blue_sc.set_alpha(0.42)
        gray_sc.set_alpha(0.18 + 0.20 * ease_in_out((t - 0.25) / 0.25))
    elif phase == 3:
        blue_sc.set_alpha(0.35)
        gray_sc.set_alpha(0.34)
    else:
        blue_sc.set_alpha(0.33)
        gray_sc.set_alpha(0.32)

    if phase == 4:
        k_ring = ease_in_out((t - 0.78) / 0.22)
        n_ring = max(1, int(len(x_mp) * k_ring))
        market_sc.set_offsets(np.column_stack([x_mp[:n_ring], y_mp[:n_ring]]))
        market_sc.set_alpha(0.25 + 0.55 * k_ring)
    else:
        market_sc.set_offsets(np.empty((0, 2)))
        market_sc.set_alpha(0.0)

    if phase == 1:
        k_ml = ease_in_out(t / 0.25)
        x_ml = x_model0
        y_ml = y_model0
        ml_alpha = 0.08 + 0.08 * k_ml
    elif phase == 2:
        k_ml = ease_in_out((t - 0.25) / 0.25)
        x_ml = interpolate(x_model0, x_model0 * 0.65, k_ml)
        y_ml = interpolate(y_model0, y_model0 * 0.65, k_ml)
        ml_alpha = 0.16 + 0.08 * k_ml
    elif phase == 3:
        k_ml = ease_in_out((t - 0.50) / 0.28)
        x_ml = interpolate(x_model0 * 0.65, x_model1, k_ml)
        y_ml = interpolate(y_model0 * 0.65, y_model1, k_ml)
        ml_alpha = 0.25 + 0.45 * k_ml
    else:
        k_ml = ease_in_out((t - 0.78) / 0.22)
        x_ml = x_model1
        y_ml = y_model1
        ml_alpha = 0.78 + 0.12 * k_ml

    ml_sc.set_offsets(np.column_stack([x_ml, y_ml]))
    ml_sc.set_alpha(ml_alpha)

    blue_count_text.set_text(f"Blue points remaining: {len(blue_now)}")
    gray_count_text.set_text(f"Gray points detected: {len(gray_now)}")

    return (
        blue_sc, gray_sc, ltv_sc, pv_sc, rs_sc, bias_sc,
        market_sc, ml_sc, epoch_text, progress_text, loss_text,
        blue_count_text, gray_count_text, progress_bg, progress_fg
    )

# =========================================================
# ANIMACIÓN
# =========================================================
ani = FuncAnimation(
    fig,
    update,
    frames=N_FRAMES,
    init_func=init,
    interval=1000 / FPS,
    blit=False
)

# =========================================================
# MÁRGENES
# =========================================================
fig.subplots_adjust(
    left=0.03,
    right=0.985,
    top=0.96,
    bottom=0.15
)

# =========================================================
# GUARDAR
# =========================================================
ani.save(OUTPUT_FILE, writer=HTMLWriter(fps=FPS))
print(f"Saved as: {OUTPUT_FILE}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, HTMLWriter
from matplotlib.lines import Line2D

# =========================================================
# CONFIGURACIÓN GENERAL
# =========================================================
plt.style.use("dark_background")
np.random.seed(123)

OUTPUT_FILE = "ml_market_animation_left_panel.html"
FPS = 20
N_FRAMES = 190

# =========================================================
# FUNCIONES AUXILIARES
# =========================================================
def ellipse_points(cx, cy, n, a, b, angle_deg=0):
    theta = np.random.uniform(0, 2 * np.pi, n)
    r = np.sqrt(np.random.uniform(0, 1, n))
    x = a * r * np.cos(theta)
    y = b * r * np.sin(theta)

    ang = np.deg2rad(angle_deg)
    xr = x * np.cos(ang) - y * np.sin(ang)
    yr = x * np.sin(ang) + y * np.cos(ang)

    return cx + xr, cy + yr


def ring_points(cx, cy, n, a, b, angle_deg=0, rmin=0.86, rmax=1.0):
    theta = np.random.uniform(0, 2 * np.pi, n)
    r = np.sqrt(np.random.uniform(rmin**2, rmax**2, n))
    x = a * r * np.cos(theta)
    y = b * r * np.sin(theta)

    ang = np.deg2rad(angle_deg)
    xr = x * np.cos(ang) - y * np.sin(ang)
    yr = x * np.sin(ang) + y * np.cos(ang)

    return cx + xr, cy + yr


def ease_in_out(t):
    t = np.clip(t, 0, 1)
    return 3 * t**2 - 2 * t**3


def interpolate(a, b, t):
    return (1 - t) * a + t * b


def sort_by_local_radius(indices, x_all, y_all, cx=0.0, cy=0.0):
    r = np.sqrt((x_all[indices] - cx) ** 2 + (y_all[indices] - cy) ** 2)
    return indices[np.argsort(r)]


# =========================================================
# DESPLAZAMIENTO HORIZONTAL DEL MERCADO
# para abrir espacio a la izquierda
# =========================================================
SHIFT_X = 1.15

# =========================================================
# ESTRUCTURA BASE DEL MERCADO
# =========================================================
centers = [
    (-1.05 + SHIFT_X, 1.05),
    ( 1.05 + SHIFT_X, 1.05),
    (-1.05 + SHIFT_X,-1.05),
    ( 1.05 + SHIFT_X,-1.05)
]
angles = [15, -15, -10, 12]

# =========================================================
# 1) CAMPO GLOBAL: TODO ARRANCA AZUL
# =========================================================
n_bg = 8500
x_bg = np.random.normal(SHIFT_X, 2.55, n_bg)
y_bg = np.random.normal(0, 2.15, n_bg)

mask_bg = ((x_bg - SHIFT_X)**2 / 4.0**2 + y_bg**2 / 3.6**2) < 1.0
x_bg = x_bg[mask_bg]
y_bg = y_bg[mask_bg]

# =========================================================
# 2) SUBMERCADOS
# =========================================================
ltv_x_all, ltv_y_all = [], []
pv_x_all, pv_y_all = [], []
rs_x_all, rs_y_all = [], []
bias_x_all, bias_y_all = [], []

for (cx, cy), ang in zip(centers, angles):
    ltv_cx, ltv_cy = cx + 0.28, cy
    x_ltv, y_ltv = ellipse_points(ltv_cx, ltv_cy, 320, 0.62, 0.50, ang)
    ltv_x_all.append(x_ltv)
    ltv_y_all.append(y_ltv)

    pv_cx, pv_cy = cx - 0.34, cy + 0.30
    x_pv, y_pv = ellipse_points(pv_cx, pv_cy, 300, 0.58, 0.45, ang)
    pv_x_all.append(x_pv)
    pv_y_all.append(y_pv)

    rs_cx, rs_cy = cx - 0.34, cy - 0.30
    x_rs, y_rs = ellipse_points(rs_cx, rs_cy, 300, 0.58, 0.45, ang)
    rs_x_all.append(x_rs)
    rs_y_all.append(y_rs)

    x_bias, y_bias = ring_points(
        cx, cy,
        n=380,
        a=1.32,
        b=1.02,
        angle_deg=ang,
        rmin=0.84,
        rmax=1.00
    )
    bias_x_all.append(x_bias)
    bias_y_all.append(y_bias)

ltv_x_all = np.concatenate(ltv_x_all)
ltv_y_all = np.concatenate(ltv_y_all)
pv_x_all = np.concatenate(pv_x_all)
pv_y_all = np.concatenate(pv_y_all)
rs_x_all = np.concatenate(rs_x_all)
rs_y_all = np.concatenate(rs_y_all)
bias_x_all = np.concatenate(bias_x_all)
bias_y_all = np.concatenate(bias_y_all)

# =========================================================
# 3) ANILLO GLOBAL
# =========================================================
x_mp, y_mp = ring_points(
    SHIFT_X, 0,
    n=900,
    a=3.0,
    b=2.7,
    angle_deg=0,
    rmin=0.88,
    rmax=1.0
)

# =========================================================
# 4) CONJUNTO MAESTRO
# =========================================================
x_all = np.concatenate([x_bg, ltv_x_all, pv_x_all, rs_x_all, bias_x_all])
y_all = np.concatenate([y_bg, ltv_y_all, pv_y_all, rs_y_all, bias_y_all])

n_bg_final = len(x_bg)
n_ltv = len(ltv_x_all)
n_pv = len(pv_x_all)
n_rs = len(rs_x_all)
n_bias = len(bias_x_all)

idx_bg = np.arange(0, n_bg_final)
idx_ltv = np.arange(n_bg_final, n_bg_final + n_ltv)
idx_pv = np.arange(n_bg_final + n_ltv, n_bg_final + n_ltv + n_pv)
idx_rs = np.arange(n_bg_final + n_ltv + n_pv, n_bg_final + n_ltv + n_pv + n_rs)
idx_bias = np.arange(
    n_bg_final + n_ltv + n_pv + n_rs,
    n_bg_final + n_ltv + n_pv + n_rs + n_bias
)

# =========================================================
# 5) CANDIDATOS A RUIDO GRIS
# =========================================================
n_gray_bg = int(0.42 * len(idx_bg))
perm_bg = np.random.permutation(idx_bg)
gray_bg_candidates = perm_bg[:n_gray_bg]
r_bg = np.sqrt((x_all[gray_bg_candidates] - SHIFT_X)**2 + y_all[gray_bg_candidates]**2)
gray_bg_candidates = gray_bg_candidates[np.argsort(r_bg)]

n_gray_ltv = int(0.28 * len(idx_ltv))
n_gray_pv = int(0.28 * len(idx_pv))
n_gray_rs = int(0.28 * len(idx_rs))
n_gray_bias = int(0.18 * len(idx_bias))

perm_ltv = np.random.permutation(idx_ltv)
perm_pv = np.random.permutation(idx_pv)
perm_rs = np.random.permutation(idx_rs)
perm_bias = np.random.permutation(idx_bias)

gray_ltv_candidates = sort_by_local_radius(
    perm_ltv[:n_gray_ltv], x_all, y_all,
    cx=np.mean(ltv_x_all), cy=np.mean(ltv_y_all)
)
gray_pv_candidates = sort_by_local_radius(
    perm_pv[:n_gray_pv], x_all, y_all,
    cx=np.mean(pv_x_all), cy=np.mean(pv_y_all)
)
gray_rs_candidates = sort_by_local_radius(
    perm_rs[:n_gray_rs], x_all, y_all,
    cx=np.mean(rs_x_all), cy=np.mean(rs_y_all)
)
gray_bias_candidates = sort_by_local_radius(
    perm_bias[:n_gray_bias], x_all, y_all,
    cx=SHIFT_X, cy=0.0
)

# =========================================================
# 6) MODELO ML
# =========================================================
model_n = 420
x_model0 = np.random.normal(SHIFT_X, 2.0, model_n)
y_model0 = np.random.normal(0, 1.6, model_n)

pool_x = np.concatenate([ltv_x_all[:180], pv_x_all[:120], rs_x_all[:120], bias_x_all[:120]])
pool_y = np.concatenate([ltv_y_all[:180], pv_y_all[:120], rs_y_all[:120], bias_y_all[:120]])
idx_pool = np.random.choice(len(pool_x), size=model_n, replace=True)
x_model1 = pool_x[idx_pool]
y_model1 = pool_y[idx_pool]

# =========================================================
# 7) FIGURA
# =========================================================
fig, ax = plt.subplots(figsize=(14, 7))
fig.patch.set_facecolor("black")
ax.set_facecolor("black")

blue_sc = ax.scatter([], [], s=5, c="#00cfff", alpha=0.42, edgecolors="none", zorder=1)
ltv_sc  = ax.scatter([], [], s=34, c="yellow", alpha=0.95, edgecolors="none", zorder=4)
pv_sc   = ax.scatter([], [], s=34, c="red", alpha=0.95, edgecolors="none", zorder=4)
rs_sc   = ax.scatter([], [], s=34, c="limegreen", alpha=0.95, edgecolors="none", zorder=4)
bias_sc = ax.scatter([], [], s=36, c="#d16cff", alpha=0.96, edgecolors="none", zorder=5)
gray_sc = ax.scatter([], [], s=8, c="lightgray", alpha=0.0, edgecolors="none", zorder=15)
market_sc = ax.scatter([], [], s=22, c="#ff9f00", alpha=0.0, edgecolors="none", zorder=7)
ml_sc = ax.scatter([], [], s=22, c="white", alpha=0.0, edgecolors="none", zorder=8)

# =========================================================
# PANEL IZQUIERDO
# =========================================================
panel_x = 0.03

title_text = ax.text(
    panel_x, 0.94, "ML Market\nFormation",
    transform=ax.transAxes,
    ha="left", va="top",
    color="white", fontsize=22, fontweight="bold",
    linespacing=0.95
)

epoch_text = ax.text(
    panel_x, 0.79, "",
    transform=ax.transAxes,
    ha="left", va="top",
    color="white", fontsize=12
)

progress_text = ax.text(
    panel_x, 0.70, "",
    transform=ax.transAxes,
    ha="left", va="top",
    color="white", fontsize=11
)

loss_text = ax.text(
    panel_x, 0.60, "",
    transform=ax.transAxes,
    ha="left", va="top",
    color="white", fontsize=11
)

blue_count_text = ax.text(
    panel_x, 0.50, "",
    transform=ax.transAxes,
    ha="left", va="top",
    color="#00cfff", fontsize=11
)

gray_count_text = ax.text(
    panel_x, 0.42, "",
    transform=ax.transAxes,
    ha="left", va="top",
    color="lightgray", fontsize=11
)

# barra de progreso en coordenadas relativas
progress_bg, = ax.plot([], [], lw=10, alpha=0.26, solid_capstyle="round", zorder=20)
progress_fg, = ax.plot([], [], lw=10, alpha=0.95, solid_capstyle="round", zorder=21)
progress_bg.set_transform(ax.transAxes)
progress_fg.set_transform(ax.transAxes)

# caja visual opcional del panel
panel_box = plt.Rectangle(
    (0.015, 0.12), 0.23, 0.80,
    transform=ax.transAxes,
    facecolor=(0, 0, 0, 0.18),
    edgecolor=(1, 1, 1, 0.08),
    linewidth=1.0,
    zorder=0
)
ax.add_patch(panel_box)

# =========================================================
# EJES
# =========================================================
ax.set_xlim(-4.8, 6.0)
ax.set_ylim(-3.2, 3.0)
ax.set_aspect("equal")
ax.axis("off")

# =========================================================
# LEYENDA ABAJO
# =========================================================
legend_elements = [
    Line2D([0], [0], marker='o', color='none', markerfacecolor='lightgray', markersize=7, label='Detected Noise'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='yellow', markersize=8, label='Long-Term Volatility'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='red', markersize=8, label='Persistent Volatility'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='limegreen', markersize=8, label='Recurrent Shocks'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='#d16cff', markersize=8, label='Bias'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='#ff9f00', markersize=7, label='Market Price Ring'),
    Line2D([0], [0], marker='o', color='none', markerfacecolor='white', markersize=7, label='ML Representation'),
]

ax.legend(
    handles=legend_elements,
    loc="upper center",
    bbox_to_anchor=(0.56, -0.05),
    ncol=4,
    facecolor="black",
    edgecolor="white",
    framealpha=0.20,
    fontsize=10,
    labelcolor="white",
    handletextpad=0.45,
    columnspacing=1.2,
    borderpad=0.55
)

# =========================================================
# INIT
# =========================================================
def init():
    empty = np.empty((0, 2))
    for sc in [blue_sc, gray_sc, ltv_sc, pv_sc, rs_sc, bias_sc, market_sc, ml_sc]:
        sc.set_offsets(empty)

    epoch_text.set_text("")
    progress_text.set_text("")
    loss_text.set_text("")
    blue_count_text.set_text("")
    gray_count_text.set_text("")

    progress_bg.set_data([], [])
    progress_fg.set_data([], [])

    return (
        blue_sc, gray_sc, ltv_sc, pv_sc, rs_sc, bias_sc,
        market_sc, ml_sc, epoch_text, progress_text, loss_text,
        blue_count_text, gray_count_text, progress_bg, progress_fg
    )

# =========================================================
# UPDATE
# =========================================================
def update(frame):
    t = frame / (N_FRAMES - 1)

    if t < 0.25:
        phase = 1
    elif t < 0.50:
        phase = 2
    elif t < 0.78:
        phase = 3
    else:
        phase = 4

    if phase == 1:
        stage = f"Epoch {frame + 1:03d}\nEverything begins in blue"
    elif phase == 2:
        stage = f"Epoch {frame + 1:03d}\nThe algorithm identifies noise"
    elif phase == 3:
        stage = f"Epoch {frame + 1:03d}\nSubmarkets and bias emerge"
    else:
        stage = f"Epoch {frame + 1:03d}\nMarket ring emerges"

    epoch_text.set_text(stage)
    progress_text.set_text(f"Training progress: {int(t * 100)}%")

    fake_loss = 1.25 * np.exp(-3.4 * t) + 0.04
    loss_text.set_text(f"Loss: {fake_loss:.3f}")

    # barra
    x0, x1 = panel_x, panel_x + 0.16
    ybar = 0.655
    progress_bg.set_data([x0, x1], [ybar, ybar])
    progress_fg.set_data([x0, x0 + (x1 - x0) * t], [ybar, ybar])
    progress_bg.set_color("#444444")
    progress_fg.set_color("#00cfff")

    # ruido gris
    if phase == 1:
        k_gray_bg = 0.0
        k_gray_inner = 0.0
    elif phase == 2:
        k_gray_bg = 0.78 * ease_in_out((t - 0.25) / 0.25)
        k_gray_inner = 0.70 * ease_in_out((t - 0.25) / 0.25)
    elif phase == 3:
        k_gray_bg = 0.78 + 0.18 * ease_in_out((t - 0.50) / 0.28)
        k_gray_inner = 0.70 + 0.25 * ease_in_out((t - 0.50) / 0.28)
    else:
        k_gray_bg = 0.96
        k_gray_inner = 0.95

    n_gray_bg_now = int(len(gray_bg_candidates) * np.clip(k_gray_bg, 0, 1))
    n_gray_ltv_now = int(len(gray_ltv_candidates) * np.clip(k_gray_inner, 0, 1))
    n_gray_pv_now = int(len(gray_pv_candidates) * np.clip(k_gray_inner, 0, 1))
    n_gray_rs_now = int(len(gray_rs_candidates) * np.clip(k_gray_inner, 0, 1))
    n_gray_bias_now = int(len(gray_bias_candidates) * np.clip(0.90 * k_gray_inner, 0, 1))

    gray_now = np.concatenate([
        gray_bg_candidates[:n_gray_bg_now],
        gray_ltv_candidates[:n_gray_ltv_now],
        gray_pv_candidates[:n_gray_pv_now],
        gray_rs_candidates[:n_gray_rs_now],
        gray_bias_candidates[:n_gray_bias_now]
    ]).astype(int)

    # submercados
    if phase in [1, 2]:
        k_sub = 0.0
        k_bias = 0.0
    elif phase == 3:
        local_t = ease_in_out((t - 0.50) / 0.28)
        k_sub = local_t
        k_bias = 0.85 * local_t
    else:
        k_sub = 1.0
        k_bias = 1.0

    n_ltv_now = int(len(idx_ltv) * k_sub)
    n_pv_now = int(len(idx_pv) * k_sub)
    n_rs_now = int(len(idx_rs) * k_sub)
    n_bias_now = int(len(idx_bias) * k_bias)

    ltv_now = idx_ltv[:n_ltv_now]
    pv_now = idx_pv[:n_pv_now]
    rs_now = idx_rs[:n_rs_now]
    bias_now = idx_bias[:n_bias_now]

    recolored = np.concatenate([gray_now, ltv_now, pv_now, rs_now, bias_now]).astype(int)

    is_recolored = np.zeros(len(x_all), dtype=bool)
    is_recolored[recolored] = True
    blue_now = np.where(~is_recolored)[0]

    blue_sc.set_offsets(np.column_stack([x_all[blue_now], y_all[blue_now]]))
    ltv_sc.set_offsets(np.column_stack([x_all[ltv_now], y_all[ltv_now]]))
    pv_sc.set_offsets(np.column_stack([x_all[pv_now], y_all[pv_now]]))
    rs_sc.set_offsets(np.column_stack([x_all[rs_now], y_all[rs_now]]))
    bias_sc.set_offsets(np.column_stack([x_all[bias_now], y_all[bias_now]]))
    gray_sc.set_offsets(np.column_stack([x_all[gray_now], y_all[gray_now]]))

    if phase == 1:
        blue_sc.set_alpha(0.44)
        gray_sc.set_alpha(0.0)
    elif phase == 2:
        blue_sc.set_alpha(0.42)
        gray_sc.set_alpha(0.18 + 0.20 * ease_in_out((t - 0.25) / 0.25))
    elif phase == 3:
        blue_sc.set_alpha(0.35)
        gray_sc.set_alpha(0.34)
    else:
        blue_sc.set_alpha(0.33)
        gray_sc.set_alpha(0.32)

    if phase == 4:
        k_ring = ease_in_out((t - 0.78) / 0.22)
        n_ring = max(1, int(len(x_mp) * k_ring))
        market_sc.set_offsets(np.column_stack([x_mp[:n_ring], y_mp[:n_ring]]))
        market_sc.set_alpha(0.25 + 0.55 * k_ring)
    else:
        market_sc.set_offsets(np.empty((0, 2)))
        market_sc.set_alpha(0.0)

    if phase == 1:
        k_ml = ease_in_out(t / 0.25)
        x_ml = x_model0
        y_ml = y_model0
        ml_alpha = 0.08 + 0.08 * k_ml
    elif phase == 2:
        k_ml = ease_in_out((t - 0.25) / 0.25)
        x_ml = interpolate(x_model0, x_model0 * 0.65, k_ml)
        y_ml = interpolate(y_model0, y_model0 * 0.65, k_ml)
        ml_alpha = 0.16 + 0.08 * k_ml
    elif phase == 3:
        k_ml = ease_in_out((t - 0.50) / 0.28)
        x_ml = interpolate(x_model0 * 0.65, x_model1, k_ml)
        y_ml = interpolate(y_model0 * 0.65, y_model1, k_ml)
        ml_alpha = 0.25 + 0.45 * k_ml
    else:
        k_ml = ease_in_out((t - 0.78) / 0.22)
        x_ml = x_model1
        y_ml = y_model1
        ml_alpha = 0.78 + 0.12 * k_ml

    ml_sc.set_offsets(np.column_stack([x_ml, y_ml]))
    ml_sc.set_alpha(ml_alpha)

    blue_count_text.set_text(f"Blue points remaining: {len(blue_now)}")
    gray_count_text.set_text(f"Gray points detected: {len(gray_now)}")

    return (
        blue_sc, gray_sc, ltv_sc, pv_sc, rs_sc, bias_sc,
        market_sc, ml_sc, epoch_text, progress_text, loss_text,
        blue_count_text, gray_count_text, progress_bg, progress_fg
    )

# =========================================================
# ANIMACIÓN
# =========================================================
ani = FuncAnimation(
    fig,
    update,
    frames=N_FRAMES,
    init_func=init,
    interval=1000 / FPS,
    blit=False
)

# =========================================================
# MÁRGENES
# =========================================================
fig.subplots_adjust(
    left=0.02,
    right=0.99,
    top=0.97,
    bottom=0.16
)

# =========================================================
# GUARDAR
# =========================================================
ani.save(OUTPUT_FILE, writer=HTMLWriter(fps=FPS))
print(f"Saved as: {OUTPUT_FILE}")
```
