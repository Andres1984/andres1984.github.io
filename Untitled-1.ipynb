```python
import numpy as np

class DiscreteHMM:
    """
    HMM discreto:
    - N estados ocultos
    - M símbolos observables
    Parámetros:
    - pi: distribución inicial, shape (N,)
    - A: matriz de transición, shape (N, N)
    - B: matriz de emisión, shape (N, M)
    """

    def __init__(self, A, B, pi, obs_symbols=None):
        self.A = np.array(A, dtype=float)
        self.B = np.array(B, dtype=float)
        self.pi = np.array(pi, dtype=float)

        self.N = self.A.shape[0]
        self.M = self.B.shape[1]

        if obs_symbols is None:
            self.obs_symbols = list(range(self.M))
        else:
            self.obs_symbols = obs_symbols

        self._check_shapes()
        self._normalize_params()

    def _check_shapes(self):
        assert self.A.shape == (self.N, self.N), "A debe ser NxN"
        assert self.B.shape == (self.N, self.M), "B debe ser NxM"
        assert self.pi.shape == (self.N,), "pi debe ser de longitud N"

    def _normalize_params(self):
        self.A = self.A / self.A.sum(axis=1, keepdims=True)
        self.B = self.B / self.B.sum(axis=1, keepdims=True)
        self.pi = self.pi / self.pi.sum()

    def forward(self, O):
        """
        Forward con scaling para estabilidad numérica.
        O: secuencia de observaciones codificadas como enteros en {0,...,M-1}
        Retorna:
        - alpha: matriz T x N
        - c: factores de scaling
        - log_likelihood
        """
        O = np.asarray(O, dtype=int)
        T = len(O)

        alpha = np.zeros((T, self.N))
        c = np.zeros(T)

        # Inicialización
        alpha[0] = self.pi * self.B[:, O[0]]
        c[0] = alpha[0].sum()
        if c[0] == 0:
            c[0] = 1e-300
        alpha[0] /= c[0]

        # Recursión
        for t in range(1, T):
            alpha[t] = (alpha[t-1] @ self.A) * self.B[:, O[t]]
            c[t] = alpha[t].sum()
            if c[t] == 0:
                c[t] = 1e-300
            alpha[t] /= c[t]

        log_likelihood = np.sum(np.log(c))
        return alpha, c, log_likelihood

    def backward(self, O, c):
        """
        Backward usando los mismos scaling factors del forward.
        """
        O = np.asarray(O, dtype=int)
        T = len(O)

        beta = np.zeros((T, self.N))
        beta[-1] = 1.0

        for t in range(T - 2, -1, -1):
            beta[t] = self.A @ (self.B[:, O[t + 1]] * beta[t + 1])
            beta[t] /= c[t + 1]

        return beta

    def viterbi(self, O):
        """
        Devuelve la secuencia de estados más probable.
        Implementado en log-espacio.
        """
        O = np.asarray(O, dtype=int)
        T = len(O)

        logA = np.log(self.A + 1e-300)
        logB = np.log(self.B + 1e-300)
        logpi = np.log(self.pi + 1e-300)

        delta = np.zeros((T, self.N))
        psi = np.zeros((T, self.N), dtype=int)

        # Inicialización
        delta[0] = logpi + logB[:, O[0]]

        # Recursión
        for t in range(1, T):
            for j in range(self.N):
                vals = delta[t - 1] + logA[:, j]
                psi[t, j] = np.argmax(vals)
                delta[t, j] = np.max(vals) + logB[j, O[t]]

        # Terminación y backtracking
        q_star = np.zeros(T, dtype=int)
        q_star[-1] = np.argmax(delta[-1])

        for t in range(T - 2, -1, -1):
            q_star[t] = psi[t + 1, q_star[t + 1]]

        best_logprob = np.max(delta[-1])
        return q_star, best_logprob, delta, psi

    def gamma_xi(self, O):
        """
        Calcula gamma y xi para Baum-Welch.
        """
        O = np.asarray(O, dtype=int)
        T = len(O)

        alpha, c, log_likelihood = self.forward(O)
        beta = self.backward(O, c)

        gamma = alpha * beta
        gamma /= gamma.sum(axis=1, keepdims=True)

        xi = np.zeros((T - 1, self.N, self.N))

        for t in range(T - 1):
            numerator = (
                alpha[t][:, None]
                * self.A
                * (self.B[:, O[t + 1]] * beta[t + 1])[None, :]
            )
            denom = numerator.sum()
            if denom == 0:
                denom = 1e-300
            xi[t] = numerator / denom

        return gamma, xi, log_likelihood

    def baum_welch(self, sequences, n_iter=50, tol=1e-6, verbose=True):
        prev_ll = -np.inf

        for it in range(n_iter):

        # ==========================================
        # ACUMULADORES DEL M-STEP
        # ------------------------------------------
        # pi_new: ocupación esperada inicial
        # A_num : transiciones esperadas i -> j
        # A_den : veces esperadas que se sale de i
        # B_num : veces esperadas que j emite símbolo k
        # B_den : ocupación esperada total del estado j
        # ==========================================
            pi_new = np.zeros(self.N)
            A_num = np.zeros((self.N, self.N))
            A_den = np.zeros(self.N)
            B_num = np.zeros((self.N, self.M))
            B_den = np.zeros(self.N)

            total_ll = 0.0

        # ==========================================
        # E-STEP
        # Para cada secuencia calculamos:
        # gamma_t(i) = P(q_t=i | O, lambda)
        # xi_t(i,j)  = P(q_t=i, q_{t+1}=j | O, lambda)
        # ==========================================
            for O in sequences:
                O = np.asarray(O, dtype=int)
                gamma, xi, ll = self.gamma_xi(O)
                total_ll += ll

                # Esperado de comenzar en cada estado
                pi_new += gamma[0]

                # Esperado de transiciones i -> j
                A_num += xi.sum(axis=0)

                # Esperado de ocupación del estado i
                # solo hasta T-1 porque desde el último estado no se transita
                A_den += gamma[:-1].sum(axis=0)

                # Esperado de emisiones:
                # si en t se observa el símbolo ot,
                # gamma[t, j] contribuye al conteo esperado de que
                # el estado j emitió ese símbolo
                for t, ot in enumerate(O):
                    B_num[:, ot] += gamma[t]

                # Ocupación total esperada por estado
                B_den += gamma.sum(axis=0)

            # ==========================================
            # M-STEP
            # Reestimamos parámetros usando conteos
            # esperados normalizados
            # ==========================================

            # Nueva distribución inicial
            self.pi = pi_new / pi_new.sum()

            # Nueva matriz de transición
            self.A = A_num / np.maximum(A_den[:, None], 1e-300)
            self.A = self.A / self.A.sum(axis=1, keepdims=True)

            # Nueva matriz de emisión
            self.B = B_num / np.maximum(B_den[:, None], 1e-300)
            self.B = self.B / self.B.sum(axis=1, keepdims=True)

            if verbose:
                print(f"Iteración {it+1:02d} | log-likelihood total = {total_ll:.6f}")

            if np.abs(total_ll - prev_ll) < tol:
                if verbose:
                    print("Convergencia alcanzada.")
                break

            prev_ll = total_ll

    def sample(self, T, random_state=None):
        """
        Genera una secuencia de estados y observaciones.
        """
        rng = np.random.default_rng(random_state)

        states = np.zeros(T, dtype=int)
        obs = np.zeros(T, dtype=int)

        states[0] = rng.choice(self.N, p=self.pi)
        obs[0] = rng.choice(self.M, p=self.B[states[0]])

        for t in range(1, T):
            states[t] = rng.choice(self.N, p=self.A[states[t - 1]])
            obs[t] = rng.choice(self.M, p=self.B[states[t]])

        return states, obs

    def log_likelihood(self, O):
        _, _, ll = self.forward(O)
        return ll

```

```python

if __name__ == "__main__":
    # ======================================================
    # EJEMPLO: clima oculto y observaciones discretas
    # Estados ocultos:
    # 0 = Soleado
    # 1 = Lluvioso
    #
    # Observaciones:
    # 0 = caminar
    # 1 = paraguas
    # 2 = café
    # ======================================================

    A = [
        [0.7, 0.3],
        [0.4, 0.6]
    ]

    B = [
        [0.6, 0.1, 0.3],   # Soleado
        [0.1, 0.7, 0.2]    # Lluvioso
    ]

    pi = [0.8, 0.2]

    hmm = DiscreteHMM(A, B, pi, obs_symbols=["caminar", "paraguas", "café"])

    # Secuencia observada
    O = [0, 2, 1, 1, 2]

    # Forward
    alpha, c, ll = hmm.forward(O)
    print("Log-likelihood:", ll)

    # Viterbi
    q_star, best_logprob, delta, psi = hmm.viterbi(O)
    print("Mejor secuencia de estados:", q_star)
    print("Log-prob mejor camino:", best_logprob)

    # Muestreo
    states, obs = hmm.sample(T=10, random_state=123)
    print("Estados simulados:", states)
    print("Observaciones simuladas:", obs)

    # Entrenamiento Baum-Welch con datos simulados
    sequences = []
    for seed in range(20):
        _, obs_seq = hmm.sample(T=15, random_state=seed)
        sequences.append(obs_seq)

    # Inicialización aleatoria para reaprender parámetros
    rng = np.random.default_rng(2026)
    A0 = rng.dirichlet(np.ones(2), size=2)
    B0 = rng.dirichlet(np.ones(3), size=2)
    pi0 = rng.dirichlet(np.ones(2))

    hmm_learn = DiscreteHMM(A0, B0, pi0)
    print("\nParámetros iniciales:")
    print("A=\n", hmm_learn.A)
    print("B=\n", hmm_learn.B)
    print("pi=\n", hmm_learn.pi)

    hmm_learn.baum_welch(sequences, n_iter=30, verbose=True)

    print("\nParámetros aprendidos:")
    print("A=\n", hmm_learn.A)
    print("B=\n", hmm_learn.B)
    print("pi=\n", hmm_learn.pi)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import FancyArrowPatch
from scipy.stats import norm

# ============================
# PARÁMETROS DEL HMM
# ============================
states = ["S1", "S2", "S3"]

# Matriz de transición
A = np.array([
    [0.70, 0.20, 0.10],
    [0.15, 0.70, 0.15],
    [0.10, 0.25, 0.65]
])

# Distribución inicial
pi = np.array([0.6, 0.3, 0.1])

# Emisiones gaussianas por estado
means = np.array([0.0, 3.0, 6.0])
stds  = np.array([0.8, 1.0, 1.2])
```

```python
def plot_transition_matrix(A, states):
    fig, ax = plt.subplots(figsize=(6, 5))
    im = ax.imshow(A, aspect='auto')

    ax.set_xticks(range(len(states)))
    ax.set_yticks(range(len(states)))
    ax.set_xticklabels(states)
    ax.set_yticklabels(states)

    ax.set_xlabel("Estado siguiente")
    ax.set_ylabel("Estado actual")
    ax.set_title("Matriz de transición")

    for i in range(A.shape[0]):
        for j in range(A.shape[1]):
            ax.text(j, i, f"{A[i, j]:.2f}", ha="center", va="center")

    plt.colorbar(im, ax=ax, label="Probabilidad")
    plt.tight_layout()
    plt.show()

plot_transition_matrix(A, states)
```

```python
def plot_state_graph(A, states):
    fig, ax = plt.subplots(figsize=(8, 6))
    ax.set_xlim(-2, 2)
    ax.set_ylim(-2, 2)
    ax.axis("off")

    positions = {
        0: (-1.2, 0.8),
        1: (1.2, 0.8),
        2: (0.0, -1.0)
    }

    # Dibujar nodos
    for i, state in enumerate(states):
        x, y = positions[i]
        circle = plt.Circle((x, y), 0.25, fill=False, linewidth=2)
        ax.add_patch(circle)
        ax.text(x, y, state, ha="center", va="center", fontsize=12)

    # Dibujar flechas
    for i in range(A.shape[0]):
        for j in range(A.shape[1]):
            if A[i, j] > 0.05:
                x1, y1 = positions[i]
                x2, y2 = positions[j]

                if i == j:
                    ax.text(x1, y1 + 0.45, f"{A[i,j]:.2f}", ha="center")
                else:
                    arrow = FancyArrowPatch(
                        (x1, y1), (x2, y2),
                        arrowstyle='->',
                        mutation_scale=15,
                        shrinkA=20,
                        shrinkB=20,
                        linewidth=1.5,
                        alpha=0.7
                    )
                    ax.add_patch(arrow)

                    mx, my = (x1 + x2) / 2, (y1 + y2) / 2
                    ax.text(mx, my, f"{A[i,j]:.2f}", fontsize=10)

    ax.set_title("Grafo de transiciones del HMM")
    plt.tight_layout()
    plt.show()

plot_state_graph(A, states)
```

```python
def state_distribution_evolution(A, pi, n_steps=15):
    distributions = [pi]
    current = pi.copy()

    for _ in range(n_steps):
        current = current @ A
        distributions.append(current)

    return np.array(distributions)

def plot_state_evolution(A, pi, states, n_steps=15):
    dist = state_distribution_evolution(A, pi, n_steps)

    fig, ax = plt.subplots(figsize=(8, 5))
    for i, state in enumerate(states):
        ax.plot(dist[:, i], marker='o', label=state)

    ax.set_xlabel("Tiempo")
    ax.set_ylabel("Probabilidad")
    ax.set_title("Evolución de la distribución de estados")
    ax.legend()
    ax.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

plot_state_evolution(A, pi, states, n_steps=20)
```

```python
def plot_emission_distributions(means, stds, states):
    x = np.linspace(-3, 10, 500)

    fig, ax = plt.subplots(figsize=(8, 5))
    for i, state in enumerate(states):
        y = norm.pdf(x, loc=means[i], scale=stds[i])
        ax.plot(x, y, label=f"{state}: μ={means[i]}, σ={stds[i]}")

    ax.set_xlabel("Observación")
    ax.set_ylabel("Densidad")
    ax.set_title("Distribuciones de emisión por estado")
    ax.legend()
    ax.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

plot_emission_distributions(means, stds, states)
```

```python
def sample_hmm(A, pi, means, stds, T=30, seed=123):
    rng = np.random.default_rng(seed)
    N = len(pi)

    hidden_states = np.zeros(T, dtype=int)
    observations = np.zeros(T)

    hidden_states[0] = rng.choice(N, p=pi)
    observations[0] = rng.normal(means[hidden_states[0]], stds[hidden_states[0]])

    for t in range(1, T):
        hidden_states[t] = rng.choice(N, p=A[hidden_states[t-1]])
        observations[t] = rng.normal(means[hidden_states[t]], stds[hidden_states[t]])

    return hidden_states, observations

hidden_states, observations = sample_hmm(A, pi, means, stds, T=40)
```

```python
def plot_hidden_and_observed(hidden_states, observations, states):
    T = len(hidden_states)
    t = np.arange(T)

    fig, axes = plt.subplots(2, 1, figsize=(10, 6), sharex=True)

    axes[0].step(t, hidden_states, where='mid')
    axes[0].set_yticks(range(len(states)))
    axes[0].set_yticklabels(states)
    axes[0].set_ylabel("Estado oculto")
    axes[0].set_title("Secuencia de estados ocultos")

    axes[1].plot(t, observations, marker='o')
    axes[1].set_xlabel("Tiempo")
    axes[1].set_ylabel("Observación")
    axes[1].set_title("Observaciones generadas por el HMM")
    axes[1].grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

plot_hidden_and_observed(hidden_states, observations, states)
```

```python
def plot_histograms_by_state(hidden_states, observations, states):
    fig, axes = plt.subplots(len(states), 1, figsize=(8, 7), sharex=True)

    for i, ax in enumerate(axes):
        obs_i = observations[hidden_states == i]
        ax.hist(obs_i, bins=12, density=True, alpha=0.7)
        ax.set_title(f"Observaciones emitidas por {states[i]}")
        ax.grid(True, alpha=0.3)

    axes[-1].set_xlabel("Valor observado")
    plt.tight_layout()
    plt.show()

plot_histograms_by_state(hidden_states, observations, states)
```
